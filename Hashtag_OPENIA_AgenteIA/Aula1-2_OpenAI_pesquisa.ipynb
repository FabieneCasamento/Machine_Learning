```python
# aula

from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

cliente = OpenAI()
```

```python
resposta = cliente.chat.completions.create() # criar um chat
# completions armazena as respostas anteriores.
```

```python
resposta = cliente.chat.completions.create(
    messages=[{"role": "system", "content": "Por que python é uma boa linguagem de programação?"}], # lista de mensagens e específico.
    model="gpt-4o"
)
```

```python
# role usário, system (linguagem), ou mensagem do usuário user


resposta = cliente.chat.completions.create(
    messages=[{"role": "user", "content": "Por que python é uma boa linguagem de programação?"}], # lista de mensagens e específico.
    model="gpt-4o"
)
print(resposta)
mensagem_resposta = resposta.choices[0].message # só a mensagem
conteudo_mensagem = mensagem_resposta.content # conteúdo da mensagem
role_mensagem = mensagem_resposta.role # quem enviou
print("Role:", role_mensagem)
print("Content:", conteudo_mensagem)
```
