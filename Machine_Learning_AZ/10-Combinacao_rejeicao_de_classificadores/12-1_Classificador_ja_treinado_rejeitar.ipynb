# 16.3 Combinação de classificadores - implementação 

Previsão para cada um dos algoritmos, e faremos combinações (como se fosse o voto da maioria)


## Combinação de classificadores

```python
novo_registro = X_credit[0]
novo_registro = novo_registro.reshape(1, -1)
print(f'registro {novo_registro}')
print(f'{novo_registro.shape}')
novo_registro
```

```python
resposta_random_forest= random_forest.predict(novo_registro)
resposta_knn = knn.predict(novo_registro)
resposta_arvore_decisao = arvore_decisao.predict(novo_registro)

print('resultado random forest:',resposta_random_forest[0]) #  0 pagaria o empréstimo
print('resultado knn:',resposta_knn[0]) # 0 pagaria
print('resultado árvore:',resposta_arvore_decisao[0]) # 0 pagaria
```

```python
resposta_random_forest
```

```python
# quanto a maior quantidade de votos para o pagaria 0 e n]ao pagaria, será a resposta final do algoritmo. 

paga = 0
nao_paga = 0

if resposta_random_forest[0] == 1:  # igual a 1 não paga, em cada algoritmo.
  nao_paga += 1
else: # igual a 0 paga
  paga += 1

if resposta_knn[0] == 1: # igual a 1 não paga, em cada algoritmo.
  nao_paga += 1
else: # igual a 0 paga
  paga += 1

if resposta_arvore_decisao[0] == 1: # igual a 1 não paga, em cada algoritmo.
  nao_paga += 1
else: # igual a 0 paga
  paga += 1

if paga > nao_paga:
  print('Cliente pagará o empréstimo')
elif paga == nao_paga: # o ideal seria usar o número impar de algoritmos, para evitar o empate. 
    # Ou estabecer pesos em porcentagens para cada algoritmo.
  print('Empate')
else:
  print('Cliente não pagará o empréstimo')
```

```python
# parte 2 teste 

novo_registro = X_credit[25]
novo_registro = novo_registro.reshape(1, -1)
print(f'registro {novo_registro}')
print(f'{novo_registro.shape}')
novo_registro

resposta_random_forest= random_forest.predict(novo_registro)
resposta_knn = knn.predict(novo_registro)
resposta_arvore_decisao = arvore_decisao.predict(novo_registro)

print('resultado random forest:',resposta_random_forest[0]) #  0 pagaria o empréstimo
print('resultado knn:',resposta_knn[0]) # 0 pagaria
print('resultado árvore:',resposta_arvore_decisao[0]) # 0 pagaria

# quanto a maior quantidade de votos para o pagaria 0 e n]ao pagaria, será a resposta final do algoritmo. 
def resposta_cliente(resposta_random_forest_0, resposta_knn_0, resposta_arvore_decisao_0):
    paga = 0
    nao_paga = 0
    
    if resposta_random_forest_0[0] == 1:  # igual a 1 não paga, em cada algoritmo.
      nao_paga += 1
    else: # igual a 0 paga
      paga += 1
    
    if resposta_knn_0[0] == 1: # igual a 1 não paga, em cada algoritmo.
      nao_paga += 1
    else: # igual a 0 paga
      paga += 1
    
    if resposta_arvore_decisao_0[0] == 1: # igual a 1 não paga, em cada algoritmo.
      nao_paga += 1
    else: # igual a 0 paga
      paga += 1
    
    if paga > nao_paga:
      print('Cliente pagará o empréstimo')
    elif paga == nao_paga: # o ideal seria usar o número impar de algoritmos, para evitar o empate. 
        # Ou estabecer pesos em porcentagens para cada algoritmo.
      print('Empate')
    else:
      print('Cliente não pagará o empréstimo')

resposta_cliente(resposta_random_forest, resposta_knn, resposta_arvore_decisao)        
```

```python
# parte 3 teste 

novo_registro = X_credit[255]
novo_registro = novo_registro.reshape(1, -1)
print(f'registro {novo_registro}')
print(f'{novo_registro.shape}')
novo_registro

resposta_random_forest= random_forest.predict(novo_registro)
resposta_knn = knn.predict(novo_registro)
resposta_arvore_decisao = arvore_decisao.predict(novo_registro)

print('resultado random forest:',resposta_random_forest[0]) #  0 pagaria o empréstimo
print('resultado knn:',resposta_knn[0]) # 0 pagaria
print('resultado árvore:',resposta_arvore_decisao[0]) # 0 pagaria
resposta_cliente(resposta_random_forest, resposta_knn, resposta_arvore_decisao)        
```

```python

# quanto a maior quantidade de votos para o pagaria 0 e n]ao pagaria, será a resposta final do algoritmo. 
def resposta_cliente_2(resposta_random_forest_0, confianca_random_forest_0, resposta_knn_0, confianca_knn_0, resposta_arvore_decisao_0, confianca_arvore_decisao_0):
    paga = 0
    nao_paga = 0
    confianca_minima = 0.999999 # confiança precisa ser alta. 
    algoritmos = 0
    
    if confianca_random_forest_0 >= confianca_minima:
      print(f'confiança no algoritmo random forest, de {confianca_random_forest_0*100:.2f}%') 
      algoritmos += 1
      if resposta_random_forest_0[0] == 1:
        nao_paga += 1
      else:
        paga += 1
    
    if confianca_knn_0 >= confianca_minima:
      print(f'confiança no algoritmo  knn, de {confianca_knn_0*100:.2f}%')        
      algoritmos += 1
      if resposta_knn_0[0] == 1:
        nao_paga += 1
      else:
        paga += 1
    
    if confianca_arvore_decisao_0 >= confianca_minima:
      print(f'confiança no algoritmo árvore de decisão, de {confianca_arvore_decisao_0*100:.2f}%.')     
      algoritmos += 1
      if resposta_arvore_decisao_0[0] == 1:
        nao_paga += 1
      else:
        paga += 1
    
    if paga > nao_paga:
      print('Cliente pagará o empréstimo, baseado em {} algoritmos'.format(algoritmos))
    elif paga == nao_paga:
      print('Empate, baseado em {} algoritmos'.format(algoritmos))
    else:
      print('Cliente não pagará o empréstimo, baseado em {} algoritmos'.format(algoritmos))



resposta_cliente_2(resposta_random_forest, confianca_random_forest, resposta_knn, confianca_knn, resposta_arvore_decisao, confianca_arvore_decisao)        
```
