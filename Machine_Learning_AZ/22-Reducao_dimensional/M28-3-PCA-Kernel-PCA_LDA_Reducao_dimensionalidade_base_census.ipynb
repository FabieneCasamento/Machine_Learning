https://scikit-learn.org/stable/modules/generated/sklearn.discriminant_analysis.LinearDiscriminantAnalysis.html



_sklearn.discriminant_analysis.LinearDiscriminantAnalysis(solver='svd', shrinkage=None, priors=None, n_components=None, store_covariance=False, tol=0.0001, covariance_estimator=None)_


* solver	

'svd'	=> Define o método usado para calcular a matriz de dispersão. As opções são:

'svd' => (Valor Padrão)	Não calcula explicitamente a matriz de covariância. Recomendado para dados com muitas features (colunas), pois é o mais robusto numericamente.

'lsqr'	=> Solução por Mínimos Quadrados. Pode ser usado com shrinkage. Mais rápido que o svd.

'eigen' => 	Decomposição por Autovalores/Autovetores. Funciona com shrinkage.

* shrinkage =>	None	Aplica a técnica de shrinkage (encolhimento) à matriz de covariância. Usada principalmente quando há poucos dados ou muitas features.

None (Padrão) => Não aplica shrinkage.
    
'auto'	=> O algoritmo calcula o melhor fator de shrinkage (apenas com solver='lsqr' ou solver='eigen').

float (0 a 1) =>	Define um fator de shrinkage constante manualmente.

* priors	None =>	Probabilidades a priori de cada classe.

None (Padrão) =>	As probabilidades são estimadas a partir da frequência relativa das classes nos dados de treinamento.

array-like	=> Uma lista de probabilidades fornecida pelo usuário (deve somar 1).

* n_components

None =>	O número de componentes que o LDA usará como dimensionalidade reduzida.

None (Padrão) =>	O valor máximo (o número de classes - 1) é usado.

int	=> Permite especificar o número de dimensões desejadas para a redução.

* store_covariance

False	=> Controla se a matriz de covariância deve ser explicitamente calculada e armazenada após o ajuste.

False (Padrão)	=> Não armazena a matriz, economizando memória.

True => 	Calcula e armazena a matriz de covariância para inspeção posterior (acessível via .covariance_).

* tol

0.0001 =>	Tolerância absoluta usada para ajudar a determinar o ranking de uma matriz de covariância. Afeta a forma como a regularização é aplicada internamente.

* covariance_estimator	

None	=> Um objeto estimator usado para estimar a matriz de covariância, como sklearn.covariance.EmpiricalCovariance.

None (Padrão) => As estimativas padrão do LDA são usadas. É usado principalmente quando se deseja uma estimativa de covariância mais robusta.

```python
random_forest_census_lda.get_params()
```

* n_estimators	40	Define que o modelo será construído a partir de 40 Árvores de Decisão individuais. O resultado final da classificação será a votação majoritária dessas 40 árvores.

* criterion	'entropy'	Define a função que mede a qualidade da divisão em cada nó da árvore. 'entropy' (Entropia) visa maximizar a informação ganha em cada divisão.

* random_state	0	Garante que o processo de criação (seleção aleatória de dados e features) seja reproduzível. Se você rodar o código novamente, obterá exatamente as mesmas árvores e o mesmo resultado.

- n_estimators	Efeito no Modelo

- Baixo (ex: 10-30)	Rápido de treinar, mas instável. O modelo pode não ter diversidade suficiente e o erro de generalização pode ser alto.

- Médio (ex: 50-200)	Bom equilíbrio. A performance geralmente se estabiliza. É o ponto onde você deve começar a testar.

- Alto (ex: 500+)	Melhor performance marginal. O ganho de performance é mínimo ou nulo, mas o tempo de treinamento e de previsão é significativamente maior.
