## Redução de dimensionalidade

### Preparacão da base de dados

```python
import pandas as pd
# material da escola, o outro do início das aulas foi "baseado em dados retirados da internet atuais".
arquivo = '../24-Dados_desbalanceados/census.csv'

base_census = pd.read_csv(arquivo)
base_census.tail()
```

```python
# dados da pessoa, para prever a renda, income
```

```python
base_census['income'].unique()
```

```python
X_census = base_census.iloc[:, 0:14].values
X_census
```

```python
#colunas de base para achar o income
list(base_census.iloc[:,0:14].columns)
```

```python
y_census = base_census.iloc[:, 14].values
y_census
```

```python
list(base_census.iloc[:, 14:15].columns)
```

## Para os atributos categóricos

```python
from sklearn.preprocessing import LabelEncoder
label_encoder_workclass = LabelEncoder()
label_encoder_education = LabelEncoder()
label_encoder_marital = LabelEncoder()
label_encoder_occupation = LabelEncoder()
label_encoder_relationship = LabelEncoder()
label_encoder_race = LabelEncoder()
label_encoder_sex = LabelEncoder()
label_encoder_country = LabelEncoder()
```

```python
# aplicação a cada um dos atributos

X_census[:,1] = label_encoder_workclass.fit_transform(X_census[:,1])
X_census[:,3] = label_encoder_education.fit_transform(X_census[:,3])
X_census[:,5] = label_encoder_marital.fit_transform(X_census[:,5])
X_census[:,6] = label_encoder_occupation.fit_transform(X_census[:,6])
X_census[:,7] = label_encoder_relationship.fit_transform(X_census[:,7])
X_census[:,8] = label_encoder_race.fit_transform(X_census[:,8])
X_census[:,9] = label_encoder_sex.fit_transform(X_census[:,9])
X_census[:,13] = label_encoder_country.fit_transform(X_census[:,13])
```

```python
X_census[0] # dos atributos categoricos para os atributos numéricos 
```

### Padronizar os dados para deixar todos na mesma escala

```python
from sklearn.preprocessing import StandardScaler
scaler_census = StandardScaler()
X_census = scaler_census.fit_transform(X_census)
```

```python
X_census[:10] # os valores foram padronizados com a mesma ordem. 
```

```python
from sklearn.model_selection import train_test_split
X_census_treinamento, X_census_teste, y_census_treinamento, y_census_teste = train_test_split(X_census, 
                                                                                              y_census, 
                                                                                              test_size=0.15, 
                                                                                              random_state=0)
```

```python
X_census_treinamento.shape, X_census_teste.shape # técnica para reduzir a dimensionalidade
```

### 14 atributos no total, e vamos reduzir a dimensionalidade, e deixar com 6 atributos apenas. 

### Vamos aplicar a classificação e fazer o comparativo dos resultados para nós verificarcmos se a aplicação do PCA vai melhorar o acurácia do algoritmo.

## PCA (Principal component analysis)

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=6) # instanciamos (objeto concreto) para número de componentes que indica quantos atributos que nós vamos gerar. 
# verificas os atributos com as datas de nascimento. Cria um novo atributo combinando a idade e o nascimento, por exemplo
# assim os atributos mais correlacionados serão UNIDOS;
```

ver os atributos mais correlacionados 

```python
# transformar os dados no x treinamento e teste
X_census_treinamento_pca = pca.fit_transform(X_census_treinamento) # algoritmo adpatou os dados


X_census_testes_pca = pca.transform(X_census_teste) # depois transforma
```

```python
X_census_treinamento_pca.shape, X_census_testes_pca.shape 

# mudança para 6 atributos, cobinando os 14 atributos e gera novos 6 atributos
```

```python
X_census_treinamento[:4]
```

```python
# verificar a variância dos dados
```

```python
pca.explained_variance_ratio_
```

```python
#14 atiributo para 1: (0.151561) teremos 15% das variáveis explicáveis 
```

```python
sum(list(pca.explained_variance_ratio_))
```

```python
sum(list(pca.explained_variance_ratio_)[:2]) # segunda coluna 25% dos dados explicado das variáveis
```

```python
sum((pca.explained_variance_ratio_))
```

```python
sum(list(pca.explained_variance_ratio_)[:3]) # terceira coluna 34% dos dados explicado das variáveis
```

```python
sum(list(pca.explained_variance_ratio_))*100 #57%  pouco ainda de interesse, explicação nas variáveis
```

```python
pca.explained_variance_ratio_.sum()
```

```python

# teste 14 atributos
# transformar os dados no x treinamento e teste
X_census_treinamento_pca = pca.fit_transform(X_census_treinamento) # algoritmo adpatou os dados
X_census_testes_pca = pca.transform(X_census_teste) # depois transforma

print(X_census_treinamento_pca.shape, X_census_testes_pca.shape)

print(X_census_treinamento[:4])

print(pca.explained_variance_ratio_)
print(sum(list(pca.explained_variance_ratio_)))
print(f' segunda coluna dos dados explicado das variáveis: {sum(list(pca.explained_variance_ratio_)[:2])} ')
print(sum(list(pca.explained_variance_ratio_)[:2])) # segunda coluna dos dados explicado das variáveis
print(sum((pca.explained_variance_ratio_)))

print(f' terceira coluna  dos dados explicado das variáveis {sum(list(pca.explained_variance_ratio_)[:3])}') # terceira coluna  dos dados explicado das variáveis

print(f'interesse total, explicação nas variáveis {sum(list(pca.explained_variance_ratio_))*100} %') # %  pouco ainda de interesse, explicação nas variáveis
print(pca.explained_variance_ratio_.sum())

#100% todas as variáveis, tem 100% de exolicação delas
```

```python

# teste para a explicação mais próxima de 1
pca = PCA(n_components=8) # 8 variáveis ou compontentes ver a porcentagem para explicar as variáveis. 
# transformar os dados no x treinamento e teste
X_census_treinamento_pca = pca.fit_transform(X_census_treinamento) # algoritmo adpatou os dados
X_census_testes_pca = pca.transform(X_census_teste) # depois transforma

print(X_census_treinamento_pca.shape, X_census_testes_pca.shape)

print(X_census_treinamento[:4])

print(pca.explained_variance_ratio_)
print(sum(list(pca.explained_variance_ratio_)))
print(f' segunda coluna dos dados explicado das variáveis: {sum(list(pca.explained_variance_ratio_)[:2])} ')
print(sum(list(pca.explained_variance_ratio_)[:2])) # segunda coluna dos dados explicado das variáveis
print(sum((pca.explained_variance_ratio_)))

print(f' terceira coluna  dos dados explicado das variáveis {sum(list(pca.explained_variance_ratio_)[:3])}') # terceira coluna  dos dados explicado das variáveis

print(f'interesse total, explicação nas variáveis {sum(list(pca.explained_variance_ratio_))*100} %') # %  pouco ainda de interesse, explicação nas variáveis
print(pca.explained_variance_ratio_.sum())

#8 variáveis ou compontentes explicam 70% das variáveis
```

### utilizar o componente de algoritmo de classificação, ao invés de nós passarmos a base de dados com 14 registros, vamos somente passar com 8 registros. 

### usar o algoritmo random forest classifier.

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
random_forest_census_pca = RandomForestClassifier(n_estimators=40, random_state=0, criterion = 'entropy')
random_forest_census_pca.fit(X_census_treinamento_pca, y_census_treinamento) # treinamento para gerar as árvores
# gera as 40 árvores
```

```python
random_forest_census_pca.get_params()
```

### gerar as previsões e verificar os resultados 


RandomForestClassifier(n_estimators=40, random_state=0, criterion='entropy')

1️⃣ n_estimators=40

Quantidade de árvores que o Random Forest vai criar.

Mais árvores → melhor desempenho, mais lento.

Aqui são 40 árvores.

2️⃣ random_state=0

Define a semente aleatória.

Serve para que o resultado seja reproduzível (igual toda vez que rodar).

3️⃣ criterion='entropy'

Define como cada árvore decide onde cortar os dados.

'entropy' → usa ganho de informação (Information Gain)

'gini' → é o padrão e costuma ser um pouco mais rápido

Aqui o escolhido foi entropy, que tende a separar melhor classes desbalanceadas, mas pode ser um pouco mais lento.

```python
# as previsões

previsoes = random_forest_census_pca.predict(X_census_testes_pca)
previsoes
```

```python
# respostas reais
y_census_teste
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_census_teste, previsoes)
```

```python
# cálculo da acurácia
print(f'acurácia do random forest {accuracy_score(y_census_teste, previsoes)*100} %')
```

## Kernel PCA

```python
from sklearn.decomposition import KernelPCA
```

```python
kpca = KernelPCA(n_components=8, kernel='rbf') # vamos passar novamente para 8 atributos, ou colunas

X_census_treinamento_kpca = kpca.fit_transform(X_census_treinamento) # treinamento
X_census_teste_kpca = kpca.transform(X_census_teste) # teste
```

# parametro adicional, kernel linear

kernel{‘linear’, ‘poly’, ‘rbf’, ‘sigmoid’, ‘cosine’, ‘precomputed’} or callable, default=’linear’

Kernel used for PCA.

```python
### rbf, problemas não linearmente separáveis
```

```python
from sklearn.ensemble import RandomForestClassifier

# número de árvores :40
# impureza de geni entropy
random_forest_census_kpca = RandomForestClassifier(n_estimators = 40, 
                                                   criterion = 'entropy', 
                                                   random_state = 0)

random_forest_census_kpca.fit(X_census_treinamento_kpca, y_census_treinamento) # treinamento
```

```python
random_forest_census_kpca.get_params()
```

🌲 n_estimators = 40

É a quantidade de árvores na floresta.
Mais árvores geralmente deixam o modelo mais estável, mas consomem mais tempo.
Com 40, você tem um equilíbrio bom entre velocidade e performance.
Número de árvores na floresta. Mais árvores → modelo mais estável (e mais lento).
40 árvores.

🧮 criterion = 'entropy'

Define como cada árvore decide onde “cortar” os dados (os splits).

entropy → usa o conceito de entropia da teoria da informação (igual no ID3).
É um critério um pouco mais exigente, buscando dividir os dados criando nós mais “puros”.

(Se fosse gini, seria um cálculo mais simples, mas ambos funcionam bem.)

Mede a qualidade da divisão usando entropia (informação).

Alternativa ao padrão "gini". Normalmente gera splits um pouco mais “puros”.


🎲 random_state = 0

Garante que o resultado seja sempre igual para rodar o código.

É como travar a semente da aleatoriedade.

Isso é ótimo para reproduzir experimento

Garante repetibilidade. Mantém os mesmos resultados se você rodar de novo.

fit(...)

Treina o modelo usando os dados transformados pelo KPCA (X_census_treinamento_kpca) e os rótulos (y_census_treinamento).
