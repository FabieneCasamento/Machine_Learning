```python
# coletar os dados no formato orange
import Orange

base_credit = Orange.data.Table('risco_credit_data_rules.csv')
# correto risco de crédito regras
base_credit
```

```python
import pandas as pd
olhar= pd.read_csv('risco_credit_data_rules.csv', sep=";") # mesmo arquivo da análise do notebook anterior
display(olhar.tail())
```

```python
base_credit.domain
```

```python
# criar o novo algoritmo
majority = Orange.classification.MajorityLearner()
majority
```

```python

# criar o novo algoritmo  MajorityLearner()

# depois ver as previsões

previsoes = Orange.evaluation.testing.TestOnTestData(base_credit, base_credit, [majority])
previsoes
```

```python
# previsoes = Orange.evaluation.testing.TestOnTestData(base_credit, base_credit, [majority])
# treinando e testando na mesma base de dados
# de fato no final é o algoritmo [majority]

# valor da acurácia
Orange.evaluation.CA(previsoes)
```

```python
acuracia = Orange.evaluation.CA(previsoes)
print(f'Acurácia do modelo {acuracia[0]*100:.2f} %')
```
