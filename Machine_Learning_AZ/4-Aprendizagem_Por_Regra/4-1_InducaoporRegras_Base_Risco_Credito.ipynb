# Base risco de crédito

##### teremos que mudar a coluna para c#risco para que o dado seja se considera como a classe. Anotação no .csv, para o origem interpretar esses dados.

```python
import pandas as pd
df_base_risco_credito = pd.read_csv('../4-Bayesiana/risco_credito.csv') 
display(df_base_risco_credito.tail())

df_base_risco_credito = df_base_risco_credito.rename(columns={'risco': 'c#risco'})
df_base_risco_credito.head()
```

```python
df_base_risco_credito.to_csv('risco_credito_regras.csv', sep=';', index= False)
```

```python
# coletar os dados no formato orange
import Orange


base_risco_credito = Orange.data.Table('risco_credito_regras.csv') # correto risco de crédito regras
base_risco_credito
```

1) o c# se separa com | , ficamos com 4 previsores e 1 classe depois do |

2) ver o domínio

```python
base_risco_credito.domain
```

https://orange.readthedocs.io/en/latest/reference/rst/Orange.classification.rules.html

```python
# cn2 algoritmo 
cn2 = Orange.classification.rules.CN2Learner() # algorimo baseado em regras
regras_risco_credito = cn2(base_risco_credito) # criando as regras
# com isso treinamento já foi concluido
```

```python
for regras in regras_risco_credito.rule_list:
  print(regras) # geradas as regras
```

```python
# história boa, dívida alta, garantias nenhuma, renda > 35
# história ruim, dívida alta, garantias adequada, renda < 15
```

```python
# história boa (0), dívida alta (0), garantias nenhuma (1), renda > 35 (2)
# história ruim (2), dívida alta (0), garantias adequada (0), renda < 15 (0)

# algoritmo consegue trabalhar com os atributos categóricos
previsoes = regras_risco_credito([['boa', 'alta', 'nenhuma', 'acima_35'], ['ruim', 'alta', 'adequada', '0_15']])
previsoes
```
