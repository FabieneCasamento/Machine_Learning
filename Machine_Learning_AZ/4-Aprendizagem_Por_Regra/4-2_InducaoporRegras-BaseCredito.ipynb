# 6.1) Indução das Regras - Base Crédito

```python
# colocaremos como i#clientid será ignorado
# será usado o income, age, loan, 
# a classe será c#default
```

```python
import pandas as pd
base_credit = pd.read_csv('../2-3-Classificacao/credit_risk_dataset.csv') 
display(base_credit.tail())
```

```python

base_credit = base_credit[['person_income', 'person_age', 'loan_amnt', 'loan_status']]

base_credit['person_income'] = base_credit['person_income'].astype('float')
base_credit['person_age'] = base_credit['person_age'].astype('float')
base_credit['loan_amnt'] = base_credit['loan_amnt'].astype('float')

base_credit = base_credit.reset_index()
base_credit = base_credit.rename(columns={"index": "clientid"})
base_credit.tail()
```

```python
# colocaremos como i#clientid que será ignorado
# será usado o person_income, person_age, loan_amnt, 
# a classe será c#default -> c#loan_status -> se a pessoa vai ou nã pagar o impréstimo 

df_base_credit = base_credit.rename(columns={'loan_status': 'c#loan_status', 'clientid': 'i#clientid'})
df_base_credit.head()
df_base_credit.to_csv('risco_credit_data_rules.csv', sep=';', index= False)
```

```python
# coletar os dados no formato orange
import Orange

base_credit = Orange.data.Table('risco_credit_data_rules.csv')
# correto risco de crédito regras
base_credit
```

```python
# ver a separação correta dos atributos
base_credit.domain
# [income, age, loan | default]
```

```python

# parâmetro n é base de dados de teste, com valor de 0.25 -> 25% para  testar e 75% para treinar
# foi criado uma matriz

base_dividida = Orange.evaluation.testing.sample(base_credit, n = 0.25)
base_dividida
```

```python
base_dividida[0] #primira parte da base de dados
```

```python
base_dividida[1] # segunda parte da base de dados
```

```python
base_treinamento = base_dividida[1]
base_teste = base_dividida[0]
```

```python
# ver o tamanho dos dados
len(base_treinamento), len(base_teste)
```

```python
cn2 = Orange.classification.rules.CN2Learner()
regras_credit = cn2(base_treinamento)  # criar as regras com base do treinamento
```

c#loan_status -> se a pessoa vai ou nã pagar o impréstimo 

Pessoa paga o empréstimo -> 0 

Pessoa NÃO PAGA O EMPRÉSTIMO -> 1



```python
# conjunto de regras gerados 
for regras in regras_credit.rule_list:
  print(regras)
```

person_income -> renda

person_age -> ano

loan_amnt -> empréstimo 



```python

# executar em sequência


```

fazer as previsões na base de teste



```python
previsoes = Orange.evaluation.testing.TestOnTestData(base_treinamento, base_teste, [lambda testdata: regras_credit])
# passar as bases de treinamento, base teste, e o tempo de execução: nome do algoritmo

# executar o teste
```

```python
previsoes # classe no formato Results
```

```python
#classification accuracy

#F1, visutalizar 

Orange.evaluation.CA(previsoes)
```

# Acurácia de 82,7% no base credit

```python

print(f'Acurácia do modelo {Orange.evaluation.CA(previsoes)[0]*100:.2f} %')
```
