```python
# criaremos as variáveis

#buscar somente a coluna 5
```

```python
X_casas = base_casas_copy.iloc[:,5:6].values   #coluna de interesse: sqft_living, metragem da casa que é o nosso previsor.
X_casas
```

```python
y_casas = base_casas_copy.iloc[:, 2].values  # valor da casa, ou price
y_casas
```

# 17.4) Regressão Linear Simples -> Base Preço Casa 1

```python
# vamos prever os preços, para isso vamos separar bases dos treinos.

from sklearn.model_selection import train_test_split
X_casas_treinamento, X_casas_teste, y_casas_treinamento, y_casas_teste = train_test_split(X_casas, y_casas, test_size = 0.3, random_state = 0)
```

```python
# 30% para testar, e os 70% para treinar.
# test_size = 0.3
# random_state=0, mesma divisão com os mesmos registros
```

```python
X_casas_treinamento.shape, y_casas_treinamento.shape
```

X => registros,  atributo previsor -> metragem quadrada

Y => y previsor, preço

```python
X_casas_teste.shape, y_casas_teste.shape
```

## Vamos criar agora o nosso modelo de previsão

```python
from sklearn.linear_model import LinearRegression
regressor_simples_casas = LinearRegression()
regressor_simples_casas.fit(X_casas_treinamento, y_casas_treinamento)
```

```python
regressor_simples_casas.get_params()
```

```python
# b0 -> parâmetro b0
regressor_simples_casas.intercept_
```

```python
# b1 -> coeficiente
regressor_simples_casas.coef_
```

```python
# previsão do novo registro
```

```python
regressor_simples_casas.score(X_casas_treinamento, y_casas_treinamento)
```

```python
qualidade_algoritmo = regressor_simples_casas.score(X_casas_treinamento, y_casas_treinamento)

print('A acurácia do modelo é de {:.1f}%'.format(qualidade_algoritmo*100)) # quanto mais perto de 1, melhor é esse valor.
```

```python
regressor_simples_casas.score(X_casas_teste, y_casas_teste)
```

```python
qualidade_algoritmo_teste =  regressor_simples_casas.score(X_casas_teste, y_casas_teste)
print('A acurácia do modelo é de {:.1f}%'.format(qualidade_algoritmo_teste*100)) # quanto mais perto de 1, melhor é esse valor.
```

Como está abaixo dos 50% o desempenho foi ruim, no exemplo. Para fazer a previsão dos preços.

```python
previsoes = regressor_simples_casas.predict(X_casas_treinamento) # coletar as metragens do treinamento para prever os preços.
previsoes
```

```python
grafico = px.scatter(x = X_casas_treinamento.ravel(), y = previsoes) # X_casas_treinamento matriz, ravel transforma em vetor.
grafico.show()
```

```python
# y = b0 +b1*x1 -> equação da reta
```

```python
# figura elegante
```

```python
grafico1 = px.scatter(x = X_casas_treinamento.ravel(), y = y_casas_treinamento) # y dados reais
grafico3 = go.Figure(data=grafico1.data)
grafico3
```

```python
grafico1 = px.scatter(x = X_casas_treinamento.ravel(), y = y_casas_treinamento) # y dados reais, gráfico pontos, dispersão
grafico2 = px.line(x = X_casas_treinamento.ravel(), y = previsoes) # y previsores, gráfico linha
grafico2.data[0].line.color = 'red'
grafico3 = go.Figure(data=grafico1.data + grafico2.data) # concatenar.
grafico3
```

# 17.6) Regressão Linear simples - Base preço casas 2

```python
previsoes_teste = regressor_simples_casas.predict(X_casas_teste) # previsão do preço, com base na métrica quadrado
previsoes_teste
```

```python
y_casas_teste # dados do price reais
```

```python
# segundo dado do preço foi feito uma previsão menor de preço. 
```

Calcular a diferença da previsão com o valor real, dados do erro. 

o ideal é ter o menor valor do erro, nos 3 casos abaixo, quanto menor o valor do erro menor é a diferença entre as previsões do algoritmo e os valores reais.

MAE

MSE penaliza erros maiores, quanto maior o erro maior será o MSE.

RMSE 



```python
y_casas_teste - previsoes_teste
```

```python
# extração do valor absoluto
abs(y_casas_teste - previsoes_teste)
```

```python
# valor absoluto, e cálcula a média.
abs(y_casas_teste - previsoes_teste).mean()  # mae mean absolute error
```

```python
# em média erra o preço de 172 mil para cima ou para a média, 
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error  # MAE e MSE
```

```python
# ideal usar o MAE, pois não ela ao quadrado como no MSE
mean_absolute_error(y_casas_teste, previsoes_teste)
```

```python
# MSE
mean_squared_error(y_casas_teste, previsoes_teste)
```

```python
# RMSE
np.sqrt(mean_squared_error(y_casas_teste, previsoes_teste))
```

```python
from sklearn.metrics import root_mean_squared_error
#root_mean_squared_error(y_true, y_pred)
root_mean_squared_error(y_casas_teste, previsoes_teste)
```

Gráfico considerando a base de teste.

```python
grafico1 = px.scatter(x = X_casas_teste.ravel(), y = y_casas_teste) # dados reais, do teste, com preço real
grafico2 = px.line(x = X_casas_teste.ravel(), y = previsoes_teste) # preço previsto, em vermelho
grafico2.data[0].line.color = 'red'
grafico3 = go.Figure(data=grafico1.data + grafico2.data)
grafico3.show()
```
