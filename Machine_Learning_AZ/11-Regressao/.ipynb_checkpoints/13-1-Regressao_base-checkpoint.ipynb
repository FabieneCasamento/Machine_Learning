### Fórmula da Covariância

$[
\text{Cov}(X, Y) = \frac{\sum_{i=1}^n (x_i - \bar{x}) \cdot (y_i - \bar{y})}{n-1}
$]

Onde:
- $( x_i $) = valor da variável X
- $( y_i $) = valor da variável Y
- $( \bar{x} $) = média da variável X
- $( \bar{y} $) = média da variável Y
- $( n $) = número de observações

```python
import numpy as np

def covariancia_manual(x, y):
    n = len(x)
    media_x = np.mean(x)
    media_y = np.mean(y)
    return np.sum((x - media_x) * (y - media_y)) / (n - 1)
```

```python
x0= [30,39,49,60]
y0 = [57000, 69000, 77000, 90000]

covariancia_manual(x0,y0)
```

### coeficiente de correlação – Cr

variação do preço é explicado pela variação do tamanho.  Cr + positiva, relação linear (indica que podemos aplicar a regressão linear).
Cr 0 não há ligação entre as variáveis

preço do café diferente do preço da casa

Cr – correlação negativa, indica uma variável aumenta a outra diminui. 

### Fórmula do Coeficiente de Correlação de Pearson (r)

$[
r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2} \cdot \sqrt{\sum (y_i - \bar{y})^2}}
$]

- $( x_i, y_i $): valores individuais das variáveis X e Y  
- $( \bar{x}, \bar{y} $): média de X e Y
- $( r $): valor do coeficiente de correlação (entre -1 e 1)

```python
import numpy as np

def correlacao_pearson(x, y):
    return np.corrcoef(x, y)[0, 1]

correlacao_pearson(x0,y0)
```

Coefience de determinação (Cr) = coeficiente de correlação ao quadrado.


98% dos dados do preço conseguem ser explicado pelo tamanho da casa. 

Quando mais próximo de 1 o cr, melhor as explicações. 

```python
Cd = (correlacao_pearson(x0,y0))**2
print(f"Cd {Cd:.2f}")
```
