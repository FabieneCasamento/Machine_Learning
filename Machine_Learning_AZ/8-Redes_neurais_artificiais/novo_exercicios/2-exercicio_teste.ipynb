```python
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, accuracy_score
import numpy as np

# Exemplo de dados
df = pd.DataFrame({
    'classe': [1, 0, 1, 0],
    'calculado': [0.300, 0.020, 0.890, 0.320]
})

# Erros
mae = mean_absolute_error(df['classe'], df['calculado'])
mse = mean_squared_error(df['classe'], df['calculado'])
rmse = np.sqrt(mse)

# Taxa de acerto (com threshold 0.5)
preditos = (df['calculado'] >= 0.5).astype(int)
acuracia = accuracy_score(df['classe'], preditos) * 100

# Exibir resultados
print(f"MAE:  {mae:.4f}")
print(f"MSE:  {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"Acurácia: {acuracia:.2f}%")
```

### Métricas de Avaliação

- **Erro Absoluto Médio (MAE)**  
  $MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$

- **Erro Quadrático Médio (MSE)**  
  $MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

- **Raiz do Erro Quadrático Médio (RMSE)**  
  $RMSE = \sqrt{MSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$

- **Acurácia (%)**  
  $Acurácia = \frac{\text{número de previsões corretas}}{\text{total de previsões}} \times 100$
