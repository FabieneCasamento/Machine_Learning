# fazer testes com menos árvores

```python
# algorítmos
from sklearn.ensemble import RandomForestClassifier

random_forest_census = RandomForestClassifier(n_estimators=10, criterion='entropy', random_state = 0)
# aletório, sempre os mesmos valores (random_state =0)
random_forest_census.fit(X_census_treinamento, y_census_treinamento)
print(random_forest_census.get_params())

# passando somente os dados de teste
previsoes = random_forest_census.predict(X_census_teste)

#from sklearn.metrics import accuracy_score
# real, previsão
acuracia = accuracy_score(y_census_teste, previsoes)
print()
print('A acurácia do modelo é de {:.2f}% para 10 árvores'.format(acuracia*100))


```

# visualizar a Mátriz de confusão 

```python
from yellowbrick.classifier import ConfusionMatrix
cm = ConfusionMatrix(random_forest_census)
cm.fit(X_census_treinamento, y_census_treinamento)
cm.score(X_census_teste, y_census_teste)
```
