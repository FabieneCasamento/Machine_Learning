### Dados binários para decimais

```python
# Número binário como string
binario_str = "11100010"

# O segundo argumento '2' diz ao Python que a string é binária (base 2), base elevada ao indice
decimal = int(binario_str, 2)

print(f"O número binário {binario_str} é igual a {decimal} em decimal.") 
```

```python
def binario_para_decimal(binario_str):
    """Converte uma string binária para um número decimal usando a lógica de potências de 2."""
    
    decimal = 0
    # Inverte a string para começar da direita (potência 2^0)
    binario_invertido = binario_str[::-1] 
    print(f'binario_invertido: {binario_invertido}')
    
    # Itera sobre cada dígito (bit)
    for indice, bit in enumerate(binario_invertido):
        # Se o bit for '1', adicione a potência de 2 correspondente ao valor decimal
        if bit == '1':
            # O índice representa a potência (0, 1, 2, 3, ...)
            valor_potencia = 2 ** indice 
            decimal += valor_potencia

            print(f'Valor potencial {valor_potencia} = 2 elevado ao indice ({indice})')
            
    return decimal
print('o valor decimal é a soma de todos os valores potenciais')
# Exemplo de uso
binario_str = "11100010"
resultado = binario_para_decimal(binario_str)

print(f"O número binário {binario_str} é igual a {resultado} em decimal.")
# Saída: O número binário 10110 é igual a 22 em decimal.
```

Cada quadradinho representa a soma do conjunto de pixels. 

Cada quadradinho há 12 pixels

Quanto maior o valor mais claro é a imagem.
