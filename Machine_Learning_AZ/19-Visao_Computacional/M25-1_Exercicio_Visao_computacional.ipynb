```python
import cv2
# carregar a imagem no notebook
image = cv2.imread('pessoas.jpg')
#cv2_imshow(image)
```

```python

import matplotlib.pyplot as plt
# Nota: cv2 lê a imagem em BGR; o Matplotlib espera RGB.

# 2. Converter de BGR para RGB para cores corretas no Matplotlib
imagem_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
fig = plt.figure(figsize=(12, 7)) 

# 3. Exibir com Matplotlib
plt.imshow(imagem_rgb)
plt.title("Imagem Exibida com Matplotlib")
plt.axis('off') # Remove os eixos
plt.show()
```

```python
# objetivo é fazer a detecção dos corpos.
```

```python
# carregar o arquivo que foi treinado para detectar o corpo das pessoas;
detector_corpo = cv2.CascadeClassifier('fullbody.xml')
# convertemos em uma imagem em preto e branco
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python

import matplotlib.pyplot as plt
# Nota: cv2 lê a imagem em BGR; o Matplotlib espera RGB.


fig = plt.figure(figsize=(12, 7)) 

# 3. Exibir com Matplotlib
plt.imshow(image_gray)
plt.title("Imagem Exibida com Matplotlib")
plt.axis('off') # Remove os eixos
plt.show()
```

```python
# Define o tamanho da figura
fig = plt.figure(figsize=(12, 7)) 

# 3. Exibir com Matplotlib, definindo o colormap para tons de cinza
plt.imshow(image_gray, cmap='gray') # <--- CORREÇÃO AQUI

plt.title("Imagem Exibida com Matplotlib (Escala de Cinza)", fontsize=14)
plt.axis('off') # Remove os eixos
plt.show()
```

```python
# método para detectar as pessoas com o corpo completo
detections = detector_corpo.detectMultiScale(image_gray, scaleFactor=1.1, minSize=(50,50))
print(len(detections))
print(detections)
for (x, y, l, a) in detections:
  cv2.rectangle(image, (x, y), (x + l, y + a), (0,255,0), 2)
#cv2_imshow(image)
```

```python
print(f'foi detectado as {len(detections)} faces') # análise pelos pixes.  x y 
```

```python
fig = plt.figure(figsize=(12, 7)) 
# detectar na imagem original
# 3. Exibir com Matplotlib
plt.imshow(image)
plt.title("Imagem Exibida com Matplotlib")
plt.axis('off') # Remove os eixos
plt.show()
```
