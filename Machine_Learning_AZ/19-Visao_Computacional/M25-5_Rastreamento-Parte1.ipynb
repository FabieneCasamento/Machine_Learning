```python
import cv2
print(cv2.__version__)
```

```python
# criar o rastreador cv2

rastreador = cv2.TrackerCSRT_create()

# rastreador mais rápido, sem delay
# rastreador = cv2.TrackerMOSSE_create()
```

```python
# leitura do vídeo 
# carregamento
video = cv2.VideoCapture('rua.mp4')

# caputrar o primeiro frame
# leu realmente
# bool ok, true o vídeo carregou bem
ok, frame = video.read()
```

```python
ok
```

# definir o balde em box, para capturar o elemento para rastrear

ROI - region of interest, região de interesse. 

```python
bbox = cv2.selectROI(frame)
print(bbox) # (285, 101, 98, 75) mostrou a posição do interesse
```

Já detectou a imagem 



```python
# selecionar o objeto e aperto espaço ou enter para rastrear, 

# se quiser rastrear clico em c
```

```python
ok = rastreador.init(frame, bbox)
print(ok) # true inicializa correto.
```

```python
while True:
   ok, frame = video.read()
   if not ok:
      break
      # termina de capturar o último frame, se for false o true para o loop infinito
      
   # atualização, acompanhar o objeto se movimentado
   ok, bbox = rastreador.update(frame)
   print(bbox)
```
