```python
def dados_imagem():
  # arquivo em .gig
  caminhos = [os.path.join('yalefaces/train/', f) for f in os.listdir('yalefaces/train/')] # ver a pasta de treinamento, listar todos os diretórios
  faces = []
  ids = []
  for caminho in caminhos:
    # Este if é caso usar o jupyter notebook do colab  
    if caminho == 'yalefaces/train/.ipynb_checkpoints':
      continue
    imagem = Image.open(caminho).convert('L') # conversão para a escala cinza.
    imagem_np = np.array(imagem, 'uint8') # a imagem será convertido em matriz np e será só número inteiro.
    id = int(os.path.split(caminho)[1].split('.')[0].replace('subject', ''))
    ids.append(id)
    faces.append(imagem_np) 
  return np.array(ids), faces
```

```python
ids, faces = dados_imagem()
```
