```python
from sklearn.tree import DecisionTreeClassifier # árvore de decisão
from sklearn.ensemble import RandomForestClassifier # Random Forest
from sklearn.neighbors import KNeighborsClassifier # KNeighbors
from sklearn.linear_model import LogisticRegression # Regressão logistica
from sklearn.svm import SVC #SVC
from sklearn.neural_network import MLPClassifier # redes neurais
# o naive base não há parâmetros relevantes para esse algoritmo
# não será usado o regra, pois a biblioteca orange é incompatível ao que será usado agora
```

# 1) árvore de decisão

```python
# rodar 30 testes, 30 testes com validações cruzadas, resultados 10 folds
# 30*10=300 teste para cada um dos algoritmos, com extração da média

resultados_arvore = []

for i in range(30):
  print(i)
  # n_splits, número de kfolds (ou número de divisões), 9 para treinamento e 1 para teste, quando o n_splits é 10
  # shuffle =algoritmo sempre misture os dados 
  #random_state= receber o valor 1, depois até 30, a cada valor teremos valores ou resultados diferentes
  kfold = KFold(n_splits=10, shuffle=True, random_state=i)
  #verificar a árvore de decisão
  #arvore = DecisionTreeClassifier(criterion='entropy', min_samples_leaf=1, min_samples_split=5, splitter='best') #aula
  # roda do teste prático 11-1_Algoritmos-backup.ipynb
  # parâmetros como foi feito os tunnings
  arvore = DecisionTreeClassifier(criterion='entropy', min_samples_leaf=10, min_samples_split=2, splitter='best')    
  #{'criterion': 'entropy', 'min_samples_leaf': 10, 'min_samples_split': 2, 'splitter': 'best'}
  #melhor acurácia em 83.070 %  
  # valores, arvores parâmetros x e y, cv cross validation (kfold dividir a base de dados
  scores = cross_val_score(arvore, X_credit, y_credit, cv = kfold)
  print('score:',scores) # scores repartido 30 vezes de 10.
  print('score média:',scores.mean())
  # adicionar os resultados médios 
  resultados_arvore.append(scores.mean())


```

```python
print(resultados_arvore) # média dos 10 scores, de 30 testes de resultados que foram divididos
```

# 2) Random Forest

```python
resultados_random_forest = []

for i in range(30):
  print(i)
  # n_splits, número de kfolds (ou número de divisões), 9 para treinamento e 1 para teste, quando o n_splits é 10
  # shuffle =algoritmo sempre misture os dados 
  #random_state= receber o valor 1, depois até 30, a cada valor teremos valores ou resultados diferentes
  kfold = KFold(n_splits=10, shuffle=True, random_state=i)
  # modelo random forest
  #random_forest = RandomForestClassifier(criterion = 'entropy', min_samples_leaf = 1, min_samples_split=5, n_estimators = 10)
  random_forest = RandomForestClassifier(criterion = 'entropy', min_samples_leaf = 1, min_samples_split=10, n_estimators = 150)
  #Parâmetros escolhod como o melhor: {'criterion': 'entropy', 'min_samples_leaf': 1, 'min_samples_split': 10, 'n_estimators': 150}
  #melhor acurácia em 84.239 %        
  scores = cross_val_score(random_forest, X_credit, y_credit, cv = kfold)
  # valores, arvores parâmetros x e y, cv cross validation (kfold dividir a base de dados
  print('score:',scores) # scores repartido 30 vezes de 10.
  print('score média:',scores.mean())
  # adicionar os resultados médios     
  resultados_random_forest.append(scores.mean())

print('-'*20)
print(resultados_random_forest)
```

# KNN

```python
resultados_knn = []

for i in range(30):
  print(i)
  # n_splits, número de kfolds (ou número de divisões), 9 para treinamento e 1 para teste, quando o n_splits é 10
  # shuffle =algoritmo sempre misture os dados 
  #random_state= receber o valor 1, depois até 30, a cada valor teremos valores ou resultados diferentes
  kfold = KFold(n_splits=10, shuffle=True, random_state=i)

  #knn = KNeighborsClassifier()
  knn = KNeighborsClassifier(n_neighbors= 20,p=1)    
  #Parâmetros escolhod como o melhor: {'n_neighbors': 20, 'p': 1}
  #melhor acurácia em 83.960 %    
  scores = cross_val_score(knn, X_credit, y_credit, cv = kfold)
  # valores, arvores parâmetros x e y, cv cross validation (kfold dividir a base de dados
  print('score:',scores) # scores repartido 30 vezes de 10.
  print('score média:',scores.mean())
  # adicionar os resultados médios       
  resultados_knn.append(scores.mean())


print('-'*20)
print(resultados_knn)
```

# Resultados logisticas - Regressão Logisticas

```python
resultados_logistica = []

for i in range(30):
  print(i)
  # n_splits, número de kfolds (ou número de divisões), 9 para treinamento e 1 para teste, quando o n_splits é 10
  # shuffle =algoritmo sempre misture os dados 
  #random_state= receber o valor 1, depois até 30, a cada valor teremos valores ou resultados diferentes
  kfold = KFold(n_splits=10, shuffle=True, random_state=i)
  #  Parâmetros escolhod como o melhor: {'C': 1.0, 'solver': 'sag', 'tol': 0.0001}
  #  melhor acurácia em 79.460 %    
  #logistica = LogisticRegression(C = 1.0, solver = 'lbfgs', tol = 0.0001)
  logistica = LogisticRegression(C = 1.0, solver = 'sag', tol = 0.0001)

  scores = cross_val_score(logistica, X_credit, y_credit, cv = kfold)
  # valores, arvores parâmetros x e y, cv cross validation (kfold dividir a base de dados
  print('score:',scores) # scores repartido 30 vezes de 10.
  print('score média:',scores.mean())
  # adicionar os resultados médios    
  resultados_logistica.append(scores.mean())

print('-'*20)
print(resultados_logistica)
```

# Rede neural

```python
resultados_rede_neural = []


for i in range(30):
  print(i)
  # n_splits, número de kfolds (ou número de divisões), 9 para treinamento e 1 para teste, quando o n_splits é 10
  # shuffle =algoritmo sempre misture os dados 
  #random_state= receber o valor 1, depois até 30, a cada valor teremos valores ou resultados diferentes
  kfold = KFold(n_splits=10, shuffle=True, random_state=i)

    
  #rede_neural = MLPClassifier(activation = 'relu', batch_size = 56, solver = 'adam')
  rede_neural = MLPClassifier(activation = 'logistic', batch_size = 56, solver = 'adam')
  #  Parâmetros escolhod como o melhor: {'activation': 'logistic', 'batch_size': 56, 'solver': 'adam'}
  #  melhor acurácia em 81.093 %    
  scores = cross_val_score(rede_neural, X_credit, y_credit, cv = kfold)
  # valores, arvores parâmetros x e y, cv cross validation (kfold dividir a base de dados
  print('score:',scores) # scores repartido 30 vezes de 10.
  print('score média:',scores.mean())
  # adicionar os resultados médios        
  resultados_rede_neural.append(scores.mean())
    
```
