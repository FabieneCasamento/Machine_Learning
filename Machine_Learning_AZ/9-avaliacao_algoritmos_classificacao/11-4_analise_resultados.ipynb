# 15.3 Teste de normalidade nos resultados ->> Teste de algoritmos

## Teste de normalidade nos resultados

- Shapiro: https://en.wikipedia.org/wiki/Shapiro%E2%80%93Wilk_test

```python


# melhor média no caso, maior -> 0.843371 Random forest	, segundo o 0.838756 do knn

leitura_resultados.describe()
```

```python
alpha = 0.05# confiabilidade do teste será de 95%
```

```python
#importar a biblioteca shapiro
from scipy.stats import shapiro
```

```python
leitura_resultados
```

```python
print('árvore')
shapiro(leitura_resultados['Arvore'])
```

```python
print('Random forest')
shapiro(leitura_resultados['Random forest'])
```

```python
print('KNN')
shapiro(leitura_resultados['KNN'])
```

```python
print('Logistica')
shapiro(leitura_resultados['Logistica'])
```

```python
print('SVM')
shapiro(leitura_resultados['SVM'])
```

```python
print('Rede neural')
shapiro(leitura_resultados['Rede neural'])
```


p_value <= alpha, os dados não são normais

Hipóteste altenativa indica que os dados não estão em uma distribuição normal. 

hipóteste nula, dados estão na distribuição normal. 

p_value >= alpha, não rejeita H0 (hipóteste nula)

p_value < alpha, rejeita H0 (hipóteste nula), hipóteste alternativa (H1) aceitável


```python
list(leitura_resultados.columns)
```

```python
for algoritmo in list(leitura_resultados.columns):

    print(algoritmo)
    valor_shapiro = shapiro(leitura_resultados['Arvore'])
    print(f'estatítico: {valor_shapiro.statistic}')
    print(f'p_value: {valor_shapiro.pvalue}')
    if valor_shapiro.pvalue >= alpha:
       print('p_value >= alpha, não rejeita H0 (hipóteste nula)')
       print(f'p_value:{valor_shapiro.pvalue} >= alpha {alpha}')
       print('aceita a hipótese nula, distribuição normal, aceitável o teste TUKEY')
    elif valor_shapiro.pvalue < alpha:
       print('p_value < alpha, rejeita H0 (hipóteste nula), e aceita a hipóteste alternativa (H1) aceitável')
       print(f'p_value:{valor_shapiro.pvalue} < alpha {alpha}')
       print('distribuição NÃO normal, fazer os testes não paramêtricos')
         
    print('*'*30)
    
    
    
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import shapiro

alpha = 0.05  # ou qualquer valor que você queira usar como nível de significância

for algoritmo in list(leitura_resultados.columns):
    print(algoritmo)
    
    # Teste de normalidade Shapiro-Wilk
    valor_shapiro = shapiro(leitura_resultados[algoritmo])
    print(f'Estatística: {valor_shapiro.statistic}')
    print(f'p_value: {valor_shapiro.pvalue}')
    
    if valor_shapiro.pvalue >= alpha:
        print('p_value >= alpha, não rejeita H0 (hipótese nula)')
        print(f'p_value: {valor_shapiro.pvalue} >= alpha: {alpha}')
        print('Aceita a hipótese nula: distribuição normal. Pode usar o teste de Tukey.')
    else:
        print('p_value < alpha, rejeita H0 (hipótese nula)')
        print(f'p_value: {valor_shapiro.pvalue} < alpha: {alpha}')
        print('Distribuição NÃO normal. Usar testes não paramétricos.')
    
    # Plot da distribuição do algoritmo
    sns.displot(leitura_resultados[algoritmo], kind='kde')
    plt.title(f'Distribuição - {algoritmo}')
    plt.xlabel(algoritmo)
    plt.ylabel('Densidade')
    plt.tight_layout()
    plt.show()

    print('*' * 40)
```
