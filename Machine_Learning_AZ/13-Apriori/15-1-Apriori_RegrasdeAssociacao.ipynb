# 20.8) Regra de associação 2

apriori(df, min_support=0.5, use_colnames=True): Esta função identifica conjuntos de itens no DataFrame dfque aparecem juntos com frequência.

dfé o DataFrame de entrada onde cada linha representa uma transação e cada coluna representa um item.

min_support=0.5especifica que apenas conjuntos de itens com suporte de pelo menos 50% devem ser considerados frequentes. Suporte é a proporção de transações que contêm o conjunto de itens.

use_colnames=Truegarante que a saída use os nomes de colunas reais dfem vez dos índices de colunas.




O objetivo deste código é encontrar combinações de itens que frequentemente ocorrem juntos no conjunto de dados, o que é útil para tarefas como análise de cesta de compras.

https://www.datacamp.com/tutorial/apriori-algorithm

```python
# chamando a função apriori, coletando as listas: transacoes
regras = apriori(transacoes)
resultados = list(regras)
# aparece as regras de associação geradas
resultados
```

```python
len(resultados) # tamanho dos resultados, muito alto, as regras estão dentro da função: ordered_statistics

# também há os support
```

```python
# chamando a função apriori, coletando as listas: transacoes

# vamos configurar melhor os parâmetros
# min_support mínimo suporte de 0,3
# min_confidence, valor da confiança será 0,8
# 
regras = apriori(transacoes, min_support = 0.3, min_confidence = 0.8)
resultados = list(regras)
print(resultados)
print('\nquantidade de dados:', len(resultados)) # quantidade de dados diminuiram
```

# 20.9) Codificar para extrair todas as regras de associação

Visualização das regras

```python
for resultado in resultados:
  #print(resultado)
  s = resultado[1] # posição 1
  print(s) # dados de suporte
  result_rules = resultado[2] # posição 2 all rules
  print(result_rules)
```

```python

for resultado in resultados:
  #print(resultado)
  s = resultado[1] # posição 1
  #print(s) # dados de suporte
  result_rules = resultado[2] # posição 2 todas as regras
  #print(result_rule) #linha da regra
  for result_rule in result_rules:
    print(result_rule)
```

```python


A = [] # "regra se"
B = [] # "regra então"
# 3 métricas abaixo
suporte = []
confianca = []
lift = []

for resultado in resultados:
  #print(resultado)
  s = resultado[1] # posição 1
  #print(s) # dados de suporte
  result_rules = resultado[2] # posição 2 todas as regras
  #print(result_rule) #linha da regra 
  for result_rule in result_rules: 
    #print(result_rule) # percorrendo as regras
    a = list(result_rule[0])
    b = list(result_rule[1])
    c = result_rule[2]
    l = result_rule[3]
    print(a, ' - ', b, ' - ', c, ' - ', l)
    A.append(a) # se
    B.append(b) # então
    suporte.append(s) # suporte geral 30%
    confianca.append(c) # confiança
    lift.append(l) # lift
```
