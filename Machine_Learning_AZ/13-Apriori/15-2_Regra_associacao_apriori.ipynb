- min_support = 0.3

Suporte (Support) mede a frequência com que um conjunto de itens (itemset) aparece em todas as suas transações.

min_support = 0.3 significa que a regra só será considerada se os itens nela aparecerem juntos em pelo menos 30% de todas as transações. É um filtro para garantir que você só trabalhe com conjuntos de itens que são realmente populares.

- min_confidence = 0.8

Confiança (Confidence) mede a probabilidade de um item ser comprado, dado que outro item já foi comprado.

 min_confidence = 0.8 significa que a regra só será considerada se houver 80% de chance de um cliente comprar o segundo item, sabendo que ele já comprou o primeiro. É um filtro para garantir a confiabilidade da regra.

- min_lift = 2

Lift (Elevação) é a métrica mais poderosa, pois ela mede a força real da associação. Ela compara a probabilidade de os itens serem comprados juntos com a probabilidade de eles serem comprados de forma independente (por acaso).

 min_lift = 2 significa que a regra só será considerada se os itens forem pelo menos duas vezes mais propensos a serem comprados juntos do que seriam por acaso. É um filtro para eliminar associações óbvias ou sem sentido, encontrando padrões verdadeiramente surpreendentes e importantes.



regras de associação que são populares (min_support), confiáveis (min_confidence) e fortes/relevantes (min_lift).

```python
from apyori import apriori
regras = apriori(transacoes, min_support = 0.3, min_confidence = 0.8, min_lift = 2)
regras
```

```python
resultados = list(regras)
len(resultados) # gerado 0 regras, base de dados muito grande
```

```python
regras = apriori(transacoes, min_support = 0.003, min_confidence = 0.8, min_lift = 2)
resultados = list(regras)
len(resultados)
# deu zero, então temos que diminuir o valor de confiança, comum em dados muito grande
```

```python
regras = apriori(transacoes, min_support = 0.003, min_confidence = 0.2, min_lift = 2)
resultados = list(regras)
len(resultados)
# diminuido o suporte e a confiança
```

```python
regras = apriori(transacoes, min_support = 0.003, min_confidence = 0.2, min_lift = 3)
resultados = list(regras)
len(resultados)
# diminuido o suporte e a confiança, e aumentar o lift, na filtragem
```

```python
A = []
B = []
suporte = []
confianca = []
lift = []

for resultado in resultados:
  s = resultado[1]
  result_rules = resultado[2]
  for result_rule in result_rules:
    a = list(result_rule[0])
    b = list(result_rule[1])
    c = result_rule[2]
    l = result_rule[3]
    A.append(a)
    B.append(b)
    suporte.append(s)
    confianca.append(c)
    lift.append(l)
```

```python
# ordenação e criar o dataframe
```

```python
rules_df = pd.DataFrame({'A': A, 'B': B, 'suporte': suporte, 'confianca': confianca, 'lift': lift})
```
