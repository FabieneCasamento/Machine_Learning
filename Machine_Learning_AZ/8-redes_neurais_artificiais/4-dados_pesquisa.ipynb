Coleta do valor absoluto do erro, sem valores negativos;


                                    Erro Quadrático Médio (MSE)
                                    A fórmula do Erro Quadrático Médio é:
                                     
                                    Onde:
                                    
                                    𝑦𝑖  = valor real
                                    
                                    𝑦^𝑖  = valor previsto
                                    
                                    𝑛 = número total de observações
                                    
                                    Se quiser também a fórmula do erro simples ao quadrado:



$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

```python
import numpy as np

# Função de erro quadrático simples
def erro_quadratico(y_real, y_previsto):
    return (y_real - y_previsto) ** 2

# Função para erro quadrático médio (Mean Squared Error - MSE)
def mean_squared_error(y_real, y_pred):
    if len(y_real) != len(y_pred):
        raise ValueError("As listas devem ter o mesmo tamanho.")
    
    erro_quadratico = 0
    for real, pred in zip(y_real, y_pred):
        #erro_quadratico += abs(real - pred) teste em aula
        erro_quadratico += (real - pred) ** 2

    mse = erro_quadratico / len(y_real)
    return mse




# Exemplo de uso:
y_reais = [0,1,1,0]
y_previstos = [0.406, 0.432, 0.437, 0.458]

# Erro quadrático para o primeiro par
print("Erro quadrático (1º valor):", erro_quadratico(y_reais[0], y_previstos[0]))

# MSE para todos os valores
print("Erro quadrático médio (MSE):", mean_squared_error(y_reais, y_previstos))

```

MSE com dados menores está conseguindo se adaptar.

Quanto maior o valor do MSE pior são os resultados da rede.

# 10.8) backpropagation, taxa de aprendizagem e momento

Estudos anteriores: Gradiente, Derivada e Cálculo do Delta, agora será o Backpropagation.

Backpropagation: Algoritmo que vai fazer a atualização nos pesos;

Parâmetros: Learning rate (taxa de aprendizagem)

Momentum (momento)

Com base nos erros, o cálculo backpropagation começa do final resultado e volta ao início para achar os novos pesos. ->  retropropagação




## Cálculo do novo peso

$$
peso_{n+1} = (peso_{n} * momento) + (entrada * delta * taxa\ de\ aprendizagem)
$$


n = peso atual

momento = parâmetro

entrada é o valor em círculo (camada oculta => aplicação da minha função da ativação - cálculo da soma, atributos)

delta = para a camada de saída e oculta

taxa de aprendizagem






error
weight

peso colocado aleatóriamente, depois decidimos a melhor rota , aí vem a descida do gradiente.  



### Parâmetros

1) Taxa de aprendizagem (learning rate)

- Define quão rápido o algoritmo vai aprender

- Alto: a convergência é rápida mas pode perde o mínimo global. 

- Baixo: será mais lento mas tem mais chances de chegar no mínimo global

valor mais alto a rede neural vai aprender mais rápido: como 0,1 ou 0,2, altos 0,9


2) Momento (momentum)

- Escapar de mínimos locais (nem sempre funciona)

multiplicação!

- Define o quão confiável é a última alteração

- Alto: Aumenta a velocidade da convergência. 

- Baixo: pode evitar mínimos locais
