
# 18.5)  Regressao com Árvores -> Base Plano de Saúde


```python
import plotly.express as px
import plotly.graph_objects as go
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

base_plano_saude2 = pd.read_csv('plano_saude2.csv')
display(base_plano_saude2.tail())
```

```python
X_plano_saude2 = base_plano_saude2.iloc[:, 0:1].values #o x recebe os dados da idade
y_plano_saude2 = base_plano_saude2.iloc[:, 1].values # o y que desejamos prever é o do custo

print('x',X_plano_saude2) # atributo preditor, que é a idade.
print('y',y_plano_saude2) # respectivos preços. Regressão polinomial, elevados ao quadrado
```

# Árvores de decisão 

## Base plano saúde

```python
# arquivo plano formato de curva
```

```python
from sklearn.tree import DecisionTreeRegressor  # árvore de decisão
regressor_arvore_saude = DecisionTreeRegressor()
regressor_arvore_saude.fit(X_plano_saude2, y_plano_saude2)
```

```python
regressor_arvore_saude.get_params()
```

```python
#  'criterion': 'squared_error',
```

```python
previsoes = regressor_arvore_saude.predict(X_plano_saude2)
previsoes
```

```python
# previsor

# score semelhantes pois há um split
regressor_arvore_saude.score(X_plano_saude2, y_plano_saude2)

```

```python
qualidade_algoritmo = regressor_arvore_saude.score(X_plano_saude2, y_plano_saude2)

print('A acurácia do modelo Árvore com regressão é de {:.1f}%.'.format(qualidade_algoritmo*100)) 
# quanto mais perto de 1, melhor é esse valor.
```

```python
y_plano_saude2
```

```python
X_plano_saude2
```

```python
grafico = px.scatter(x = X_plano_saude2.ravel(), y = y_plano_saude2, labels={'Custo Real': 'Real'})
grafico.add_scatter(x = X_plano_saude2.ravel(), y = previsoes, name = 'Regressão')

grafico.update_layout(
        title=f'Comparação custo real vs Previsto Plano de Saúde',
        title_x=0.5
    )

grafico.show()
grafico.write_html(f"Grafico_Arvore_regressao_planosaude3.html")
```

```python
# buscando a idade mínima e máxima na base de dados, e colocando um encremento de 0,1 em 0,1
# dados de 18 até 63 anos de idade, todos os números de 0,1 a 0,1. 
X_teste_arvore = np.arange(min(X_plano_saude2), max(X_plano_saude2), 0.1)
X_teste_arvore
```

```python
# previsão do plano de saúde para todas as idades.
```

```python
X_teste_arvore.shape #450 dados de idade
```

```python
# colocar no formato de matriz
X_teste_arvore = X_teste_arvore.reshape(-1,1)
X_teste_arvore.shape # 450 linhas e 1 coluna
```

```python
previsor_algoritmo = regressor_arvore_saude.predict(X_teste_arvore)
previsor_algoritmo
```

```python
grafico = px.scatter(x = X_plano_saude2.ravel(), y = y_plano_saude2)
grafico.add_scatter(x = X_teste_arvore.ravel(), y = regressor_arvore_saude.predict(X_teste_arvore), name = 'Regressão')
grafico.update_layout(
        title=f'Comparação custo real vs Previsto Plano de Saúde - split',
        title_x=0.5
    )

grafico.show()
grafico.write_html(f"Grafico_Arvore_regressao_planosaude_split3_2.html")
```
