✅ 1. Teste de Friedman

Objetivo: Detectar se há diferença significativa entre três ou mais métodos (modelos, tratamentos, algoritmos, etc.) aplicados aos mesmos conjuntos de dados (ex: acurácias em vários datasets).

Tipo de teste: Não paramétrico (sem suposição de normalidade).

Equivalente ao ANOVA de medidas repetidas, mas para dados não normais.

✅ 2. Teste de Nemenyi

Objetivo: Usado após o Friedman, se ele indicar diferença significativa.

Compara cada par de métodos para saber quais diferem entre si.

É um teste post-hoc de comparação múltipla.

```python
#!pip install scikit-posthocs
```

 Suponha que temos as acurácias de 3 modelos em 5 datasets:

```python
import numpy as np
from scipy.stats import friedmanchisquare
import scikit_posthocs as sp

# Acurácias dos modelos em 5 datasets
modelo_A = [0.80, 0.82, 0.78, 0.75, 0.77]
modelo_B = [0.83, 0.85, 0.80, 0.79, 0.81]
modelo_C = [0.78, 0.80, 0.76, 0.74, 0.75]

# Teste de Friedman
stat, p = friedmanchisquare(modelo_A, modelo_B, modelo_C)
print(f"Estatística de Friedman: {stat:.4f}, p-valor: {p:.4f}")
```

```python
#Se p < 0.05: há diferença significativa entre os modelos.
```

Teste de Nemenyi (se p < 0.05)

```python
import pandas as pd

# Organiza os dados (cada linha = dataset, cada coluna = modelo)
dados = pd.DataFrame({
    'Modelo_A': modelo_A,
    'Modelo_B': modelo_B,
    'Modelo_C': modelo_C
})

# Teste de Nemenyi
posthoc = sp.posthoc_nemenyi_friedman(dados)
print(posthoc)
```

Interpretação:

Matriz com p-valores para cada par de modelos.

Se p < 0.05 entre dois modelos: diferença significativa entre eles.
