Combinação de no mínimo 1 produto min_combination

e combinação de no máximo 3 produtos max_combination

```python
indices, suporte = eclat.fit(min_support=0.3, min_combination=1, max_combination=3)
```

```python
indices
```

café aparece 3 vezes 'cafe': [1, 2, 3], aparece em 3 transações e os números são os locais das transações, no caso o indice do dataframe.
