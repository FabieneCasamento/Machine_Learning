```python
import pandas as pd

from pyECLAT import ECLAT
```

## Algoritmo ECLAT

Carregue a lista de compras a partir do arquivo `itemlist.txt` na pasta do Drive, como um DataFrame do Pandas, lembre-se de selecionar o delimitador adequado.

```python
base = pd.read_csv(caminho, delimiter=";", header=None)
base
```

## Instancie uma classe de ECLAT, informando o DataFrame.

```python
eclat = ECLAT(data=base) # instânciar os dados
```

## Exiba o DataFrame no formato binário.

```python
eclat.df_bin
```

```python
# 1 comprados

# 0 não comprados
```

## Exiba os valores das colunas, que representam os itens deste problema.

```python
eclat.uniq_  # nome dos produtos na nossa base de dados
```

## Ajuste o algoritmo ECLAT, utilizando o mesmo suporte mínimo do exercício do algoritmo apriori ($0.02$), e solicitando `min_combination=1` e `max_combination=3`.

# suporte de no mínimo 2% de presença.

```python
# itens mais frequentes

indices, support = eclat.fit(min_support=0.02, min_combination=1, max_combination=3)
```
