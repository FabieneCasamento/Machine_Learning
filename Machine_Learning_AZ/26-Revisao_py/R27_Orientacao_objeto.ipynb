```python
# armazenar as informações do triângulo
# base de como o triângulo é formado
class Triangulo:
  # primeira função gerada e rodada é no __init__(), ou construtor
  # triângulo com 3 lados e com base e altura
  def __init__(self, lado1, lado2, lado3, base, altura):
    # atributos  -> todos os atributos definidos
    # self com característica para o triângulo, é padrão escrever self
    # sem o self é o parâmetro normal
    self.lado1 = lado1
    self.lado2 = lado2
    self.lado3 = lado3
    self.base = base
    self.altura = altura
  
  def area(self): # área = base * altura /2
    # tem que colocar self nos atributos para funcionar  
    return (self.base * self.altura) / 2

  def tipo(self):
    if self.lado1 > self.lado2 + self.lado3:
      return 'não é um triângulo'
    elif self.lado1 == self.lado2 or self.lado1 == self.lado3 or self.lado1 == self.lado2:
      return 'triângulo isósceles'
    else:
      return 'outro tipo de triângulo'
```

```python
t1 = Triangulo(5, 1, 3, 4, 3)
print(t1.lado1, t1.lado2, t1.lado3, t1.base, t1.altura)
```

```python
# podemos definir mais de um objeto
t2 = Triangulo(8, 8, 8, 16, 9)
print(t2.lado1, t2.lado2, t2.lado3, t2.base, t2.altura)
```

t1 um objeto, t2 outro objeto, com novas características do Triangulo(...)

```python
print(f'objeto t1 : lado 1 {t1.lado1}, lado 2 {t1.lado2}, lado 3 {t1.lado3}, base {t1.base}, altura {t1.altura}')
t1.area() # ideia começar também
```

```python
print('tipo do triângulo do objeto t1')
t1.tipo()
```

```python
print(f'objeto t2: lado 1 {t2.lado1}, lado 2 {t2.lado2}, lado 3 {t2.lado3}, base {t2.base}, altura {t2.altura}')
print('área do triângulo do objeto t2')
t2.area()
```

um triângulo isósceles é um triângulo que possui dois lados de mesma medida
