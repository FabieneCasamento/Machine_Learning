## 34.0) Manipulação de strings

```python
a = 'casaco'
print(a)
```

# Maiúscula

```python
maiuscula = a.upper()
print(maiuscula)
```

# Minúscula

```python
minuscula = maiuscula.lower()
print(minuscula)
```

## Primeira maiúscula 

```python
capital = a.capitalize()
print(capital)
```

```python
capital = maiuscula.capitalize()
print(capital)
```

Metade da palavra que eu quero mexer

```python
metade_palavra = a[0:3]
print(metade_palavra)
```

```python
metade_palavra = a[0:4] # do 0 até o 3
print(metade_palavra)
```

```python
ultimas_letras = a[3:]
print(ultimas_letras)
```

## trocar caracteres 

```python
b = a.replace('aco', 'inha')
print(a)
print(b)
```

```python
c = a.replace('o', 'a')
print(c)
```
