# Tratamento de erros e exceções

- NameError: variável nao foi definida
- TypeError: tipos de dados incompatíveis
- RuntimeError: erro de execução
- SyntaxError: sintaxe digitada é inválida e não reconhecida pelo interpretador
- ZeroDivisionError: divisão por zero
- IndexError: índice está fora da coleção (comum em matrizes)

## Tratamento de erros

```python
# para não mostrar erros
try:
  n = int(input('Digite um número inteiro: '))
except:
  print('Valor inválido')
else:
  print(f'Valor digitado é {n}')
```

```python
# para não mostrar erros
try:
  n = int(input('Digite um número inteiro: '))
except:
  print('Valor inválido')
else:
  print(f'Valor digitado é {n}')
```

```python
# fazer um loop
while True:
  try:
    n = int(input('Digite um número inteiro: '))
  except:
    print('Valor inválido')
  else:
    print(f'Valor digitado é {n}')
    break # para sair do while
```

```python
while True:
  try:
    p = int(input('Digite um número inteiro'))
  except ValueError:
    print('Valor inválido') # valor inválido
  except KeyboardInterrupt:
    print('Usuário interrompeu a execução') # pausa no python
    break
  else:
    print(f'Valor digitado é {p}')
    break
```
