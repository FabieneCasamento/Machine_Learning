## Função com passagem de parâmetros e retorno

```python
def soma(a, b):
  return a + b
```

```python
soma(3, 2)
```

```python
# r = 7
r = soma(3, 2)
print(r)
```

```python
# r = 7
r = soma(a =3, b= 2)
print(r)
```

Cálculo físico da energia potencial gravitacional

```python
def calcula_energia_potencial_gravitacional(m, h, g = 10):
  '''
  Calcula a energia potencial gravitacional
  Argumentos:
  m: massa, entrada como uma variável float (kg)
  h: altura, entrada como uma variável float (m)

  Argumento opcional:
  g: aceleração gravitacional, com valor default de 10 m/s2

  Resposta em  1 J = 1kg .m²/s²
  '''
  e = g * m * h
  return e
```

m: Massa do objeto (kg)

g: Aceleração da gravidade (aprox. $\(9,8\ m/s^{2}$\) na superfície terrestre)

h: Altura ou posição em relação ao referencial (m)

```python
calcula_energia_potencial_gravitacional(30, 12)
```

```python
calcula_energia_potencial_gravitacional(m = 30, h = 12)
```

```python
# 1 J = 1kg .m²/s²
```

```python
print(f' Energia Potencial Gravitacional é {calcula_energia_potencial_gravitacional(m = 30, h = 12)}  J ou (kg .m²/s²). ')
```

```python
calcula_energia_potencial_gravitacional(30, 12, 9.8)
```

```python
print(f' Energia Potencial Gravitacional é {calcula_energia_potencial_gravitacional(m = 30, h = 12, g = 9.8)}  J ou (kg .m²/s²). ')
```
