Um Palíndromo é uma palavra, frase ou número que permanece igual quando lido de frente para trás ou de trás para frente.

A palavra vem do grego palin (novo/de novo) e dromos (percurso), significando literalmente "que volta pelo mesmo caminho".

1. Exemplos de Palíndromos
Palavras: * Ovo

Arara

Radar

Reviver

2. Frases (ignorando espaços e acentos):

"Ame o poema"

"A sacada da casa"

"Roma me tem amor"

3. Números (também chamados de Números Capicua):

121

2002

1331
