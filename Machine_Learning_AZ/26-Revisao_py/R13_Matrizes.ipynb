```python
import numpy as np
```

```python
# para funções estatísticas, inteligência artificial, matrizes
```

```python
# fazer matrizes
matriz = np.array([[2, 3, 1],
                   [4, 5, 7]])
```

```python
matriz
```

```python
matriz.shape  # a matriz tem 2 linhas e 3 colunas
```

```python
# primeira linha
matriz[0]
```

```python
# segunda linha
matriz[1]
```

```python
matriz[0][0] # linha 0 e coluna 0
```

```python
matriz[0][1] # linha 0 e coluna 1
```

```python
matriz[0][2] # linha 0 e coluna 2
```

```python
matriz[1][2] # linha 1 e coluna 2
```

```python
matriz.shape
```

```python
# rodar as matrizes
```

```python
for i in range(matriz.shape[0]): # percorrer as linhas
  print(f'i {i}')
  print(f'matriz: {matriz[i]}')
  for j in range(matriz.shape[1]): # percorrer as colunas
    #print(f'(i {i} e j {j})')
    print(f'matriz ({i},{j}): {matriz[i][j]}')
```
