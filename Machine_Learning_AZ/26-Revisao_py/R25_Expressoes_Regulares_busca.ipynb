## Função search

```python
import re

frase = 'Olá, meu número de telefone é (42)0010-0000'

# encontrar o telefone dentro de uma string
re.search('\(\d{2}\)\d{4,5}-\d{4}', frase)
```

```python
import re

frase = 'Olá, meu número de telefone é (42)00010-0000'

# encontrar o telefone dentro de uma string
re.search('\(\d{2}\)\d{4,5}-\d{4}', frase)
```

```python
frase2 = """Olá, meu número de telefone é (42)00010-0000, (20)00010-0050, outro
       (32)00010-0030

       """
# encontrar o telefone dentro de uma string
# d é digito, posição achada span=(30, 44), primeiro valor
re.search('\(\d{2}\)\d{4,5}-\d{4}', frase2)
```

```python
print(re.search('\(\d{2}\)\d{4,5}-\d{4}', frase2))
```

```python
frase = 'A placa de carro que eu anotei durante o acidente foi FrT-1998'
# padrão letras 3 e depois do - achar números em 4 espaços
re.search('[A-Za-z]{3}-\d{4}', frase)
```

```python
# \w caracteres + para mais caracteres
# domínio @
# \w caracteres + para mais caracteres
# busca pelo . => \.
# busca do com no final

email = 'Entre em contato, meu email é teste@teste.com'
re.search('\w+@\w+\.com', email)
```

```python
# dúvidas, busca mais dados
import re

frase2 = """Olá, meu número de telefone é (42)00010-0000, (20)00010-0050, outro
       (32)00010-0030
       """

# findall retorna uma lista com todas as strings que combinam com o padrão
telefones = re.findall('\(\d{2}\)\d{4,5}-\d{4}', frase2)

print(telefones)
# Saída: ['(42)00010-0000', '(20)00010-0050', '(32)00010-0030']
```

### a função search busca o primeiro caracter da busca.

### Já o findall busca a lista

## Função match

função match busca no início do texto

```python
frase = 'A placa de carro que eu anotei durante a batida foi FRT-1998'

print(re.match('[A-Za-z]{3}-\d{4}', frase))
```

```python
frase2 = 'FRT-1998 é a placa do carro'
print(re.match('[A-Za-z]{3}-\d{4}', frase2))
```

```python
print(re.match('[A-Z]{3}-\d{4}', frase2))
```

## Função findall

Busca mais valores;

```python
frase3 = 'Meu número de telefone atual é (42)0000-0000. O número (56)1111-1111 é o antigo'
re.findall('\(\d{2}\)\d{4,5}-\d{4}', frase3)
```

```python
emails = '''Nome: Teste 1
email: teste1@teste.com
Nome: Teste 2
email: teste2@teste.com
Nome: Teste 3
email: teste3@teste.com.br
'''
```

```python
# \w + mais de um caracter
# @ valor desejado
#\. ponto
# \w* não necessariamente seja .com
#* é nenhuma ou mais de uma vez (no caso apareceu mais .com
re.findall('\w+@\w+\.\w*', emails)
```

```python
re.findall('[\w.+-]+@[\w-]+\.[\w.-]+', emails) # atual 
```

```python
re.findall('\w+@\w+\.\w+', emails)
```

```python
re.findall('\w+@\w+\.\w-+', emails)
```

```python
re.findall('\w+@\w+\.[\w.-]+', emails) # teste
```

```python
re.findall('\w+@\w+\.\w+.?\w+', emails) #Colaboração/sugestão
```
