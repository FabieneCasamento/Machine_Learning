## 34.9) Estruturas de repetição - while

```python
# comando enquanto => while
```

```python
numero = 1
while numero < 6:
  print(numero)
  numero += 1 # esta parte é para não entrar em um loop infinito
  #numero = numero +1
print('---')
```

```python
numero = 5
while numero > 0:
  print(numero)
  numero -= 1 # incremento para diminuir o valor e não ser um loop infinito
```

```python
1 + 2 + 3 + 4 + 5
```

```python
soma = 0
numero = 1
while numero < 6:
  print(f'numero: {numero} + soma {soma}')  
  soma += numero
  numero += 1
  print(f'soma: {soma}')  
    
print(soma)
```

```python
# para digitar o dado

numero = 12
while numero < 1 or numero > 10: 
  numero = int(input('Digite um número de 1 a 10: ')) 
# tem que colocar o valor os números dentro desse limite, caso contrário, processa novamente
```
