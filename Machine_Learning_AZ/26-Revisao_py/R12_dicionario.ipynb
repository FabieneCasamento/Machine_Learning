# 35.3) Dicionários

```python
# dicionário como chave e valor
```

```python
# mosquitos, como o da dengue, 
coleta = {'Aedes aegypt': 32, 'Aedes albopictus': 22, 'Anopheles darlingi': 14}
```

```python
coleta['Aedes aegypt']
```

```python
# coletando uma nova espécie, e adicionando o novo registro
coleta['Rhodnius montenegrensis'] = 11
```

```python
print(coleta)
```

```python
# deletar uma chave em específico
del(coleta)['Aedes albopictus']
print(coleta)
```

```python
# visualizar o dicionário completo, chave  + valores
coleta.items()
```

```python
# a chave
coleta.keys()
```

```python
# valores
coleta.values()
```

```python
# juntar dois dicionários
coleta2 = {'Anopheles gambiae': 13, 'Anopheles deaneorum': 14}
print(coleta2)
```

atualizar o dicionário

```python
coleta.update(coleta2)
print(coleta)
```

```python
coleta.items()
```

```python
# percorrer o dicionário
for especie, num_especimes in coleta.items():
  print(f'Espécie: {especie}, número de espécimes coletados: {num_especimes}')
```

## Conjuntos (set)

```python
biomeleculas = ('proteína', 'ácidos nucleicos', 'carboidrato', 'lipídeo',
                'ácidos nucleicos', 'carboidrato', 'carboidrato', 'carboidrato')
```

```python
print(biomeleculas)
```

```python
# set só traz o valor único, não se repete
```

```python
biomeleculas
```

```python
print(set(biomeleculas)) # elemento que não se repete, único elemento
```

Intersecção é o ponto ou área onde duas ou mais coisas se cruzam, unem ou têm elementos em comum.

```python
c1 = {1,2,3,4,5}
c2 = {3,4,5,6,7}
c3 = c1.intersection(c2)
```

```python
c3
```

```python
# teoria de conjunto
```

```python
print(c3)
```

Cálculo da diferença

```python
# está no conjunto 1 mas não no conjunto 2
c1.difference(c2)
```

```python
# está no conjunto 2 mas não no conjunto 1
c2.difference(c1)
```
