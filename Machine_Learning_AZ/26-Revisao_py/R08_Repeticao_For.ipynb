# 34.8) Estruturas de repetição - for

```python
# for para não fazer desta forma:
print(1)
print(2)
print(3)
print(4)
print(5)
```

```python
# range é faixa
# range (vai, um a mais que printa)
for numero in range(1, 6):
  print(numero)
```

```python
# range é faixa
# range (vai (começa), um a mais que printa, step de quanto em quanto imprimi)
# indo de -1 em -1
for numero in range(5, 0, -1):
  print(numero)
```

```python
# quero fazer este cálculo dentro de um for
5 + 4 + 3 + 2 + 1
```

```python
soma = 0
for numero in range(1, 6):
  print(f'número: {numero} => (cálculo: {soma} + {numero})')  
  soma = soma + numero
  print(f'número: {numero} | somar: {soma}')
  print()

print(soma)
```

```python
palavra = 'sorvete'
for letra in palavra:
  print(letra)
```

```python
palavra = 'sorvete'
for letra in palavra:
  print(letra)
  if letra == 'v':
    print('Achou a letra v')
```

```python
# ir de 0 até 4
for i in range(0,5):
  print(i)
  print('---')
  # 0 até 2
  for j in range(0,3):
    print(j)
  print()
```
