# Exercícios

Crie expressões regulares para extrair as seguintes informações do texto abaixo (use a função findall):
- Números
- CEPs
- URLs


```python
texto = "Minha casa fica na rua Carneiro, 78. O CEP é 88388-000 e 11111-111 meu site é https://www.iaexpert.academy http://iaexpert.com.br"
```

```python
# para encontrarmos números
import re
re.findall('\d', texto)
```

```python
# formato de lista, todos os valores de números
```

```python
import re
# D maiúsculo é todos os caracteres, sem ser números
re.findall('\D', texto)
```

```python
import re
#  w todos os caracteres alpha numéricos
re.findall('\w', texto)
```

```python
import re
# W todos os caracteres que não são letras
re.findall('\W', texto)
```

```python
# para retornar o cep
# números \d{5} com 5 digitos
#- traço normal
#\d{3} número com 3 digitos
re.findall('\d{5}-\d{3}', texto)
```

```python
#? significa zero ou uma vez
# será tanto http ou https  no => https?
# padrão url ://
#/[A-Za-z0-9] varios nomes ou números
# padrão ponto -> porém era \.
# + uma ou mais vezes
re.findall('https?://[A-Za-z0-9./]+', texto)
```

```python
re.findall('https?://[A-Za-z0-9\.]+', texto)
```
