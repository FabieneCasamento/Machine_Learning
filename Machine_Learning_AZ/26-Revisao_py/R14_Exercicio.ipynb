# Exercício 1 - listas

Crie uma estrutura de repetição para fazer a leitura de 5 números inteiros e os armazene dentro de uma lista. Após a leitura, crie outra estrutura de repetição para somar todos os valores digitados

```python
lista = []
for _ in range(1, 6):
  valor = int(input('Digite o valor: '))
  lista.append(valor)

print(lista)
```

```python
len(lista)
```

```python
soma = 0
for i in range(len(lista)):
  print(f'{i}) Valor da lista: {lista[i]}')
  soma += lista[i]
  print(f'soma: {soma}')
print('Soma: ', soma)
```

# Exercícios 2 - dicionários

Crie um dicionário para armazenar o nome e a nota de 3 alunos, fazendo a leitura dos valores por meio de uma estrutura de repetição. Depois, crie uma nova estrutura de repetição para somar todas as notas e retornar a média

```python
alunos = {}
for _ in range(1, 4):
  nome = input('Digite o nome: ')
  nota = float(input('Digite a nota: '))
  alunos[nome] = nota
```

```python
alunos
```

```python
alunos['paula']
```

```python
# apenas a soma dos valores
alunos.values()
```

```python
soma = 0
for nota in alunos.values():
  print(nota)
  soma += nota
print('Média: ', soma / 3)
```

```python
soma = 0
for nome, nota in alunos.items():
  print(nome, ':',nota)
  soma += nota
print(f'Média: {(soma / 3):.2f}')
```

# Exercício 3: Matriz

Dada a matriz abaixo, construa uma estrutura de repetição para percorrer e somar todos os elementos da matriz

```python
# criar a matriz
matriz = np.array([[3, 4, 1],
                   [3, 1, 5]])
matriz
```

```python
matriz.shape
```

```python
print(f'Linhas {matriz.shape[0]} e colunas {matriz.shape[1]} na matriz.')
```

```python
soma = 0
for i in range(matriz.shape[0]): # linhas
  for j in range(matriz.shape[1]): # colunas
    print(f'matriz em ({i},{j}): {matriz[i][j]} e sua soma com: {soma}')
    soma += matriz[i][j]
    print(f'Total da soma: {soma}')

print('Soma final: ', soma)
```
