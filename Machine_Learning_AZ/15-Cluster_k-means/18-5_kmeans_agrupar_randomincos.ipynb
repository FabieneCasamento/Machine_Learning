# 22.4) Dados randômicos

```python
# geração dos dados aleatórios
from sklearn.datasets import make_blobs

```

```python
# na criação estipulamos a quantidade de registros 200, números dos cluster 5 e fixar as rodadas random_state =1
X_random, y_random = make_blobs(n_samples=200, centers=5, random_state=1)
# para fixar random_state =1, quando sempre rodarmos dará sempre o mesmo valor. Mesma disposição dos dados. 
```

```python
# conjunto de valores aleatórios
X_random
```

```python
y_random # 0 até 4 são os critérios e onde cada grupo representa, se pertence a qual cluster.
```

```python
# gráfico para ver a disposição dos dados
```

```python
# importar as bibliotecas
import plotly.express as px
# concatenar os dados
import plotly.graph_objects as go
import numpy as np
```

```python
grafico = px.scatter(x = X_random[:,0], y = X_random[:,1], title='<b>Gráfico aleatórios x vs y</b>')

# Atualiza o layout para renomear os rótulos dos eixos X e Y
grafico.update_layout(
    xaxis_title_text='X_random x',
    yaxis_title_text='X_random y',
    # padrão o título
    title_font_size=20,  # Define o tamanho da fonte do título
    title_font_family='Arial Black', # Uma fonte mais grossa
    title_x=0.5 # Centraliza o título no gráfico
)

grafico.show()
grafico.write_html("figuras/18-5_grafico1.html")
#https://plotly.com/python/interactive-html-export/
```

# queremos achar o centroide


```python
from sklearn.cluster import KMeans
```

```python
kmeans_blobs = KMeans(n_clusters=5) # já sabemos que há 5 grupos
kmeans_blobs.fit(X_random)
```

```python
kmeans_blobs.get_params()
```

```python
len(X_random)
```

```python
import math
# Use a função sqrt() da biblioteca math
resultado_num_cluster = math.sqrt(len(X_random)/2)
print(resultado_num_cluster) # ideal a quantidade de cluster
```

# achar os rotulos pelo atributo labels ou pode ser predict

mesmos valores

```python
rotulos = kmeans_blobs.labels_
rotulos
```

```python
# atual valor
rotulos = kmeans_blobs.predict(X_random)
rotulos
```

```python
# CRIAR Matriz com 5 posições, onde há 5 clusters

centroides = kmeans_blobs.cluster_centers_
centroides

```

```python


grafico1 = px.scatter(x = X_random[:,0], y = X_random[:,1], color = rotulos) # posição original, com rotulos para verificar as cores
grafico2 = px.scatter(x = centroides[:,0], y = centroides[:,1], size = [5, 5, 5, 5, 5]) # valores da média de cada grupo(cluster), centroides
grafico3 = go.Figure(data = grafico1.data + grafico2.data) # concatenar os valores
# Atualiza o layout para renomear os rótulos dos eixos X e Y
grafico3.update_layout(
    # Defina o título aqui, no gráfico que será exibido
    title_text='<b>Gráfico aleatórios x vs y</b>',
    xaxis_title_text='X_random x',
    yaxis_title_text='X_random y',
    # padrão o título
    title_font_size=24,  # Define o tamanho da fonte do título
    title_font_family='Arial Black', # Uma fonte mais grossa
    title_x=0.5 # Centraliza o título no gráfico
)
grafico3.show()
grafico3.write_html("figuras/18-5_grafico2.html")
```
