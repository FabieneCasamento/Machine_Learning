### Distância Euclidiana

$$
DE(x, y) = \sqrt{\sum_{i=1}^{n} (x_i - y_i)^2}
$$


ou

$$
DE(x, y) = \sqrt{\sum_{i=1}^{p} (x_i - y_i)^2}
$$

#### Cálculo da distância euclidiana

```python
# 1 verificação da distância:

x_verde = [21,2900]
y_vermelho =[37,1850]
x= x_verde
y= y_vermelho

import numpy as np

def distancia_euclidiana(x, y):
    return np.sqrt(np.sum((np.array(x) - np.array(y))**2))
print('distância euclidiana:')
print(distancia_euclidiana(x, y))  
print('-'*20)

#subtração de cada posição do vetor
# converte para arrays NumPy
print('subtração de cada posição do vetor')
sub = np.array(x) - np.array(y)
print(sub)  
# elevação ao quadrado
print('elevação ao quadrado')
elev= np.array(sub)**2
print(elev)

# somatória
print('somatória da elevação')
soma=np.sum(elev)
print(soma)
# raiz quadrada:
print('Raiz quadrada ~ Distância Euclidiana:')
raiz_quadrada_distancia_euclidiana = np.sqrt(soma)
print(f'Distância Euclidiana do verde para o vermelho: {raiz_quadrada_distancia_euclidiana:.2f}')
```
