# 8.1) Regressão Logística - Base Census

```python
import pickle
# read =rb 
with open('../2-3-Classificacao/census_aula.pkl', 'rb') as f:
  X_census_treinamento, y_census_treinamento, X_census_teste, y_census_teste = pickle.load(f)
```

```python
X_census_treinamento.shape, y_census_treinamento.shape
```

```python
X_census_teste.shape, y_census_teste.shape
```

```python
from sklearn.linear_model import LogisticRegression

logistic_census = LogisticRegression(random_state = 1) # random, para ter o mesmo resultado
logistic_census.fit(X_census_treinamento, y_census_treinamento)
```

```python
# verificar o parâmetro utilizado

print(logistic_census.get_params())
```

```python
previsoes = logistic_census.predict(X_census_teste)
previsoes[:50] # previsões do algoritmo
```

```python
y_census_teste[:50]
```

```python
from sklearn.metrics import accuracy_score, classification_report
accuracy_score(y_census_teste, previsoes)
```

```python
acuracia = accuracy_score(y_census_teste, previsoes)
print(f'Acurácia da Regressão Logística {acuracia*100:.2f} %')
```

```python
print('equivale ao parâmetro b0 =  {}'.format(logistic_census.intercept_[0]))

print(f'achar os coeficientes: {logistic_census.coef_[0]}')
```

Matriz de confusão

```python
from yellowbrick.classifier import ConfusionMatrix
cm = ConfusionMatrix(logistic_census)
cm.fit(X_census_treinamento, y_census_treinamento)
cm.score(X_census_teste, y_census_teste)
```

```python
print(classification_report(y_census_teste, previsoes))
```
