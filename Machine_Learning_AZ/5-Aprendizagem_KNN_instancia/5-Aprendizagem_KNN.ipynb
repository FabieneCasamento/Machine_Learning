### KNN é um método não-paramétrico que não constrói um modelo fixo. Ele utiliza todos os dados do treinamento para fazer previsões, considerando as similaridades dos vizinhos durante a classificação.


### as técnicas de normalização e padronização ajustam os dados para que compartilhem uma escala comum, o que é crucial para algoritmos que dependem de medidas de distância ou magnitude.

### a normalização, os valores são ajustados para um intervalo entre 0 e 1, enquanto na padronização os dados são transformados com base na média e no desvio padrão, garantindo escalas apropriadas para análise em algoritmos de aprendizado de máquina.

### Coeficiente de Pearson não mede distâncias, mas sim a correlação entre variáveis, enquanto o kNN é baseado em métricas de distância como a Euclidiana ou Manhattan.
