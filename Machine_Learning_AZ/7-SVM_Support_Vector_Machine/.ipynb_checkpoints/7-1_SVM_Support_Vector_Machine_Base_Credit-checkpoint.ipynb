```python
# kernel é o mais simples -> kernel='linear', random_state=1 -> para gerar sempre o mesmo valor
# canto maior o C melhor e perfeita a divisão. Se menor, pode ter resultados ruins
#C= punição por classificação incorreta
svm_credit = SVC(kernel='linear', random_state=1, C = 1.0) # default c=1, lembrando que o c o processo é mais demorado.
```

```python
svm_credit.fit(X_credit_treinamento, y_credit_treinamento)
# fit atributos previsores
```

```python
svm_credit.get_params()
```

```python
previsoes = svm_credit.predict(X_credit_teste) # passando os registros de teste
```

```python
previsoes[:500]
```

```python
y_credit_teste[:500] # respostas reais
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
accuracy_score(y_credit_teste, previsoes)
```

```python
acuracia = accuracy_score(y_credit_teste, previsoes)
print(f'Acurácia da SVM - Support Vector Machine {acuracia*100:.2f} %')
```

```python
from yellowbrick.classifier import ConfusionMatrix
```

```python
cm = ConfusionMatrix(svm_credit)
cm.fit(X_credit_treinamento, y_credit_treinamento)
cm.score(X_credit_teste, y_credit_teste)
print(classification_report(y_credit_teste, previsoes))
```

recall corretamente, e com precisão

```python

def teste_kernel(kernel_0,C_0):
    svm_credit = SVC(kernel=f'{kernel_0}', random_state=1, C =C_0) # default c=1
    svm_credit.fit(X_credit_treinamento, y_credit_treinamento)
    print(svm_credit.get_params())
    previsoes = svm_credit.predict(X_credit_teste) # passando os registros de teste
    display(previsoes[:200])
    display(y_credit_teste[:200]) # respostas reais
    acuracia = accuracy_score(y_credit_teste, previsoes)
    print(f'Acurácia da SVM - Support Vector Machine {acuracia*100:.2f} %, para Kernel: {kernel_0} e C: {C_0}')  

    cm = ConfusionMatrix(svm_credit)
    cm.fit(X_credit_treinamento, y_credit_treinamento)
    cm.score(X_credit_teste, y_credit_teste)
    
    print(classification_report(y_credit_teste, previsoes))
    return svm_credit, previsoes, cm

```

```python
# # testar para melhorar a acurácia
# tentar o kernel 'poly' adicionar mais dados na nossa base de dados, pode elevar ao quadrado.
#svm_credit = SVC(kernel='poly', random_state=1, C = 1.0) # 2 -> 4

svm_credit, previsoes, cm = teste_kernel('poly',1.0)

```

```python
svm_credit, previsoes, cm  = teste_kernel('sigmoid',1.0)
```

```python
svm_credit, previsoes, cm = teste_kernel('rbf',1.0)
```

```python
# fazer o teste com o C
svm_credit, previsoes, cm = teste_kernel('rbf',2.0)
```

o valor C muito maior, o algoritmo não vai conseguir fazer a atualização eficiente, podendo chegar ao seu limite e dar os mesmos resultados que o valor menor. Ou seja, não pode ser muito maior o C.

```python
# fazer o teste com o C
svm_credit, previsoes, cm = teste_kernel('rbf',3.0)
```

```python
# fazer o teste com o C
svm_credit, previsoes, cm = teste_kernel('rbf',5.0)
```

```python
# fazer o teste com o C
svm_credit, previsoes, cm = teste_kernel('rbf',7.0) #12:30 início # lembrar de shut down all kernel
```
