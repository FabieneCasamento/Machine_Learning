```python
# importar as bibliotecas

from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
from yellowbrick.classifier import ConfusionMatrix

#recall corretamente, e com precisão

```

```python

def teste_kernel(kernel_0,C_0):
    svm_census = SVC(kernel=f'{kernel_0}', random_state=1, C =C_0) # default c=1
    svm_census.fit(X_census_treinamento, y_census_treinamento)
    print(svm_census.get_params())
    previsoes = svm_census.predict(X_census_teste) # passando os registros de teste
    display(previsoes[:200])
    display(y_census_teste[:200]) # respostas reais
    acuracia = accuracy_score(y_census_teste, previsoes)
    print(f'Acurácia da SVM - Support Vector Machine {acuracia*100:.2f} %, para Kernel: {kernel_0} e C: {C_0}')  

    cm = ConfusionMatrix(svm_census)
    cm.fit(X_census_treinamento, y_census_treinamento)
    cm.score(X_census_teste, y_census_teste)
    
    print(classification_report(y_census_teste, previsoes))
    return svm_census, previsoes, cm

```

```python
svm_credit, previsoes, cm = teste_kernel('linear',1.0) # default c=1, lembrando que o c o processo é mais demorado.
```

(<=50) 94% dos dados detectados (recall), com precisão de 87%
