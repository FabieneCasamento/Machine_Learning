```python
import psutil

# Obtém informações sobre a memória virtual (RAM)
memoria = psutil.virtual_memory()

# psutil.virtual_memory() retorna os valores em bytes (B)
memoria_total_bytes = memoria.total

# Conversão para Gigabytes (GB) para facilitar a leitura
# 1 GB = 1024 * 1024 * 1024 bytes (aproximadamente 10^9)
memoria_total_gb = memoria_total_bytes / (1024**3)

print(f"Capacidade total de RAM (em bytes): {memoria_total_bytes} B")
print(f"Capacidade total de RAM (em GB): {memoria_total_gb:.2f} GB")

# Exemplo de como ver a memória disponível no momento
memoria_disponivel_gb = memoria.available / (1024**3)
print(f"Memória RAM disponível: {memoria_disponivel_gb:.2f} GB")
```
