```python
def buscar_token(arquivo):
    with open(arquivo, 'r') as f:
        for linha in f:
            if 'HF_TOKEN' in linha:
                # Divide no '=' e remove espaços ou quebras de linha
                return linha.split('=')[1].strip()

# Uso:
meu_token = buscar_token('env')
#print(meu_token) # senha token
```

```python
from huggingface_hub import InferenceClient
from dotenv import load_dotenv

load_dotenv()
```

```python

# dizer qual é o meu modelo
# bom modelo de text Generation
#cliente = InferenceClient(model="meta-llama/Llama-3.2-3B-Instruct")
cliente = InferenceClient(token=meu_token, model="meta-llama/Llama-3.2-3B-Instruct")
```

```python
## arquivo nomeado com ohfapi_textgeneration.py

prompt = "What is Python programming language?"

# resposta = cliente.text_generation(prompt)

# print(resposta)

# Use a pipeline as a high-level helper
# from transformers import pipeline

# pipe = pipeline("text-generation", model="meta-llama/Llama-3.2-3B-Instruct")
# messages = [
#     {"role": "user", "content": "Who are you?"},
# ]
# pipe(messages)

# Em vez de cliente.text_generation(...)
resposta = cliente.chat_completion(
    messages=[
        {"role": "user", "content": "What is Python programming language?"}
    ],
    model="meta-llama/Llama-3.2-3B-Instruct",
    max_tokens=500
)

print(resposta.choices[0].message.content)
```
