### baixar o modelo pipeline pré treinado

```python
# só pelos modelos diffusers
from diffusers import StableDiffusionPipeline
import torch

model_id = "sd-legacy/stable-diffusion-v1-5"
```

Se a sua internet estiver oscilando, você pode tentar forçar o pipeline a ser mais "paciente" ou usar uma versão simplificada do comando de download. Tente adicionar o parâmetro resume_download=True:

```python
# torch.float16, criar matrizes com vários números, gerando imagens
# tipo de dados float 32 padrão 32 bits, números pesados, 
# número mais leve e ocupa a menor a memória do computador, com o float16
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
#pipe = pipe.to("cuda")
```

```python
#pipe = pipe.to("cupu")
```

```python
prompt = "a photo of an astronaut riding a horse on mars"
resposta = pipe(prompt)
print(resposta)

imagem = resposta.images[0]

imagem.save("imagem7_0.png")
```

```python
prompt = "a purple cow with a big smile"
resposta = pipe(prompt)
print(resposta)

imagem = resposta.images[0]

imagem.save("imagem7_1.png")
```
