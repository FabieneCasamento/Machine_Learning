2. Construa seu circuito
   
Abra um ambiente Python e, em seguida, execute este código para criar um estado Bell (dois qubits emaranhados).

https://quantum.cloud.ibm.com/docs/pt/guides/quick-start

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
 
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()
 
sampler = StatevectorSampler()
result = sampler.run([qc], shots=1024).result()
print(result[0].data.meas.get_counts())
```

3. Visualize seus resultados

Para obter um histograma de seus resultados, adicione o seguinte código ao seu programa.

```python
# Uncomment lines 2 and 8 if you are not using Python in a Jupyter notebook
# import matplotlib.pyplot as plt
from qiskit.visualization import plot_histogram
 
counts = result[0].data.meas.get_counts()
plot_histogram(counts)
 
# plt.show()
```
