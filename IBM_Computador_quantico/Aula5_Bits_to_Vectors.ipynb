#bits_to_vectors

bits to vectors in python

```python
import numpy as np
# SymPy: Biblioteca para álgebra, cálculo e matemática simbólica.
# 'as sp': Um apelido (alias) curto para não precisar digitar 'sympy' toda hora.
import sympy as sp
```

O que o SymPy faz de especial?

Cálculo Exato: Ele resolve frações e raízes sem perder precisão com casas decimais.

Álgebra: pode criar variáveis $x$, $y$, $z$ e pedir para ele expandir equações como $(x + y)^2 = x^2 + 2xy + y^2$.

Cálculo Diferencial e Integral: Ele resolve derivadas e integrais complexas e te mostra a fórmula resultante.

Resolução de Equações: Ele encontra o valor de $x$ em equações como $x^2 + 5x + 6 = 0$.

```python
# Definindo 'x' como um símbolo matemático
x = sp.symbols('x')

# Criando uma expressão
expressao = (x + 2)**2

# Pedindo para o SymPy expandir (resolver o produto notável)
resultado = sp.expand(expressao)
print(resultado) # Saída: x**2 + 4*x + 4
```

## Voltando à computação quântica

```python
ket_0 = np.array([ [1],
                   [0]
                  ])
print(ket_0)
```

```python
sp.Matrix(ket_0)
```

```python
ket_1 = np.array([ [0],
                   [1]
                  ])
print(ket_1)


sp.Matrix(ket_1)

```

```python
# raiz quadrada da soma ao quadrado
np.sqrt(np.sum(ket_0**2))
```

```python
ones = np.array([ [1],
                  [1]
                  ])
print(ones)

print()
display(sp.Matrix(ones))

# raiz quadrada da soma ao quadrado
np.sqrt(np.sum(ones**2))
```

```python
# dot product
np.vdot(ket_0,ket_0)
```

```python
np.sqrt(np.vdot(ket_0,ket_0))
```

```python
np.sqrt(np.vdot(ones,ones))
```

```python
np.vdot(ket_0,ket_1)
```

```python
ket_10 = np.kron(ket_1,ket_0)
sp.Matrix(ket_10)
```
