Fazendo "na mão" (Divisões Sucessivas)

13 ÷ 2 = 6, sobra 1

6 ÷ 2 = 3, sobra 0

3 ÷ 2 = 1, sobra 1

1 ÷ 2 = 0, sobra 1

```python
numero = 13
restos = []

while numero > 0:
    resto = numero % 2      # Pega o que sobra da divisão
    restos.append(resto)    # Guarda na lista
    print(f"{numero} dividido por 2 dá {numero // 2} e sobra {resto}")
    numero = numero // 2    # Atualiza o número para a próxima divisão

# Inverte a lista para ler de baixo para cima
binario = "".join(map(str, restos[::-1]))
print(f"\nO resultado final é: {binario}")
```

```python
numero = 13

# 1. Forma padrão (com prefixo 0b)
print(bin(numero))        # Saída: '0b1101'

# 2. Apenas o número (usando format)
print(format(numero, 'b')) # Saída: '1101'

# 3. Com 8 bits (completando com zeros à esquerda)
print(f"{numero:08b}")     # Saída: '00001101'
```
