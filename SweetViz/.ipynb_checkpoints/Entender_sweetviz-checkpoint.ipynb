```python
import pandas as pd
import sweetviz as sv

import plotnine as pn
```

## dataset

```python
mpg_df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv")
```

```python
mpg_df.tail()
```

```python
# sweetviz

report = sv.analyze(
    mpg_df,
    target_feat = 'mpg'
    )
    
```

```python
report.show_html(
     
     filepath= "report_my.html"
     
     
     )
    
```
