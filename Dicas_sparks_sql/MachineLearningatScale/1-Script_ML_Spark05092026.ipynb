O que é o Databricks Feature Store?
O Databricks Feature Store (gerenciado via FeatureEngineeringClient) é um repositório centralizado dentro do Databricks onde você salva, organiza e compartilha os seus recursos (features) preparados para Machine Learning.

Em vez de salvar os dados processados em uma tabela genérica, registrar como Feature Table traz vantagens essenciais para MLOps:

Reutilização: Permite que outros cientistas de dados usem os mesmos recursos calculados sem recalcular do zero.

Prevenção de Data Leakage: Garante junções automáticas baseadas em tempo e chaves primárias na hora do treino/inferência.

Rastreabilidade (Lineage): Mantém a linhagem entre o modelo e os recursos utilizados.

```python
from pyspark.ml.recommendation import ALS
from pyspark.sql.types import StructType, StructField, IntegerType, FloatType

# 1. Criar dados de exemplo (userId, movieId, rating)
schema = StructType([
    StructField("userId", IntegerType(), True),
    StructField("movieId", IntegerType(), True),
    StructField("rating", FloatType(), True)
])

data = [
    (1, 101, 5.0), (1, 102, 3.0), (1, 103, 2.5),
    (2, 101, 2.0), (2, 102, 5.0), (2, 104, 4.5),
    (3, 102, 4.0), (3, 103, 4.5), (3, 104, 1.0),
    (4, 101, 4.0), (4, 104, 5.0)
]

ratings = spark.createDataFrame(data, schema)

# 2. Configurar o ALS com tratamento para dados frios/ausentes (coldStartStrategy)
als = ALS(
    maxIter=10, 
    regParam=0.1, 
    userCol="userId", 
    itemCol="movieId", 
    ratingCol="rating",
    coldStartStrategy="drop"  # Evita valores NaN em predições de novos usuários/itens
)

# 3. Treinar o modelo
model = als.fit(ratings)

# 4. Gerar as 2 melhores recomendações de filmes para TODOS os usuários
# Nota: recommendForAllUsers usa funções de ordem superior não suportadas no Unity Catalog
# Alternativa: gerar predições para todas combinações user-movie e selecionar top 2 por usuário
from pyspark.sql import Window
from pyspark.sql.functions import row_number, col

# Criar todas as combinações de usuário-filme
users = ratings.select("userId").distinct()
movies = ratings.select("movieId").distinct()
user_movie_pairs = users.crossJoin(movies)

# Gerar predições para todas as combinações
predictions = model.transform(user_movie_pairs)

# Selecionar top 2 filmes por usuário (maiores predições de rating)
window_spec = Window.partitionBy("userId").orderBy(col("prediction").desc())
userRecs = predictions.withColumn("rank", row_number().over(window_spec)) \
    .filter(col("rank") <= 2) \
    .select("userId", "movieId", "prediction", "rank")
display(userRecs)

# 5. Gerar os 2 usuários mais prováveis de gostar para um subconjunto de filmes (ex: filmes 101 e 102)
movies_subset = ratings.select("movieId").filter("movieId in (101, 102)").distinct()
user_movie_subset = users.crossJoin(movies_subset)
predictions_subset = model.transform(user_movie_subset)

window_spec_movies = Window.partitionBy("movieId").orderBy(col("prediction").desc())
movieSubsetRecs = predictions_subset.withColumn("rank", row_number().over(window_spec_movies)) \
    .filter(col("rank") <= 2) \
    .select("movieId", "userId", "prediction", "rank")
display(movieSubsetRecs)
```

```python
from pyspark.ml.feature import StringIndexer, OneHotEncoder, Imputer, StandardScaler, VectorAssembler
from pyspark.ml import Pipeline

# 1. Criar dados de exemplo com nulos, categóricos e numéricos
data = [
    (1, "Dog", "Black", 3.0, 30.0),
    (2, "Cat", "Brown", None, 10.0),  # Idade nula para testar o Imputer
    (3, "Dog", "White", 2.0, 20.0),
    (4, "Bird", "Black", 5.0, 5.0)
]
columns = ["AnimalID", "AnimalType", "color", "Age", "Weight"]
df_animals = spark.createDataFrame(data, columns)

# 2. Imputer: Trata valores nulos na coluna 'Age'
imputer = Imputer(
    inputCols=["Age"], 
    outputCols=["imputedAge"]
).setStrategy("mean")

# 3. StringIndexer + OneHotEncoder: Converte texto em numérico e cria vetores esparsos
type_indexer = StringIndexer(inputCol="AnimalType", outputCol="AnimalTypeIndex", handleInvalid="keep")
type_encoder = OneHotEncoder(inputCol="AnimalTypeIndex", outputCol="AnimalTypeVec")

# 4. VectorAssembler: Agrupa as features numéricas para alimentar o StandardScaler
assembler_for_scaling = VectorAssembler(
    inputCols=["imputedAge", "Weight"], 
    outputCol="unscaled_features"
)

# 5. StandardScaler: Padroniza as colunas numéricas (média 0, desvio padrão 1)
scaler = StandardScaler(
    inputCol="unscaled_features", 
    outputCol="scaled_features", 
    withMean=True, 
    withStd=True
)

# 6. Agrupar tudo em um Pipeline e transformar os dados
pipeline = Pipeline(stages=[
    imputer, 
    type_indexer, 
    type_encoder, 
    assembler_for_scaling, 
    scaler
])

pipeline_model = pipeline.fit(df_animals)
transformed_df = pipeline_model.transform(df_animals)

# Exibir resultado final
display(transformed_df.select(
    "AnimalID", "AnimalType", "AnimalTypeVec", "Age", "imputedAge", "Weight", "scaled_features"
))
```

O BinaryClassificationEvaluator suporta duas métricas principais via metricName:

"areaUnderROC": Área sob a curva ROC (AUC-ROC), padrão para verificar capacidade de discriminação geral.

"areaUnderPR": Área sob a curva Precision-Recall, útil para datasets desbalanceados.

Abaixo está o exemplo calculando ambas as métricas diretamente no seu conjunto de predições de diabetes:

```python
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# 1. Supondo que você já gerou o DataFrame 'test_predictions' com o modelo
# test_predictions = pipeline_model.transform(test_df)

# 2. Avaliador para Area Under ROC (AUC-ROC)
evaluator_roc = BinaryClassificationEvaluator(
    labelCol="diabetes",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC",
)

roc_auc = evaluator_roc.evaluate(test_predictions)
print(f"Area Under ROC (AUC): {roc_auc:.4f}")

# 3. Avaliador para Area Under Precision-Recall (PR-AUC)
evaluator_pr = BinaryClassificationEvaluator(
    labelCol="diabetes",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderPR",
)

pr_auc = evaluator_pr.evaluate(test_predictions)
print(f"Area Under Precision-Recall: {pr_auc:.4f}")
```
