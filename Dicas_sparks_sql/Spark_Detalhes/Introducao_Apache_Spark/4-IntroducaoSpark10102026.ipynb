## 1. DataFrameReader and DataFrameWriter

The DataFrame API provides two dedicated interfaces for data I/O:

* **DataFrameReader** (accessed via `spark.read`):
  * Supports multiple formats (CSV, JSON, Parquet, ORC, Delta, and more)
  * Provides schema inference and customizable options (headers, delimiters, inferSchema, etc.)

* **DataFrameWriter** (accessed via `df.write`):
  * Writes DataFrames back to external storage or tables
  * Supports various save modes (`append`, `overwrite`, `errorIfExists`, `ignore`) and partitioning options


Every DataFrame has a defined schema — the structure and data types of all its columns. A schema can be inferred from the data or explicitly specified (which is **more efficient**):

* Self-describing formats like Parquet and Delta include schema information, so no inference is needed.
* Use the `printSchema()` method to print out a DataFrame's schema.

### Schema Inference Costs
Inferring a schema requires an extra pass over the data and can guess wrong (for example, reading a postal code or ID as an integer instead of a string). For production pipelines, **always define schemas explicitly**.

```python
%python
# Defining an explicit schema for better performance and safety
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

#/Volumes/workspace/default/airbnb_data
#/Volumes/workspace/default/airbnb_data/airbnb-cleaned-mlflow.csv
schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True)
])

# Reading a CSV file applying the explicit schema
df_explicit = (
    spark.read
    .format("csv")
    .option("header", "true")
    .schema(schema)
    .load("/Volumes/workspace/default/airbnb_data/airbnb-cleaned-mlflow.csv")
)

# Inspect the DataFrame's structure
df_explicit.printSchema()
display(df_explicit.show(10))
```

```python
# Reading a CSV file applying the explicit schema
df_explicit = (
    spark.read
    .format("csv")
    .option("header", "true")
    #.schema(schema)
    .load("/Volumes/workspace/default/airbnb_data/airbnb-cleaned-mlflow.csv")
)

# Inspect the DataFrame's structure
df_explicit.printSchema()
```

```python
display(df_explicit.show(10))
```

### DDL Schemas

Instead of defining schemas using `StructType` and `StructField` Python objects, you can use **DDL (Data Definition Language) strings**. DDL strings provide a much more readable and concise syntax for schema definitions:

* Formatted as a single string using standard SQL data types (e.g., `STRING`, `INT`, `DOUBLE`, `TIMESTAMP`).
* Supports modifiers like `NOT NULL`.

```python
%python
# Defining a schema using a DDL string
ddl_schema = "name STRING NOT NULL, age INT, city STRING"

# Using the DDL schema in the DataFrameReader
df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .schema(ddl_schema)
    .load("/Volumes/workspace/default/airbnb_data/people.csv")
)

# Print out the resulting DataFrame structure
df.printSchema()
```

```python
df = (
    spark.read
    .format("csv")
    .option("header", "true")
    #.schema(ddl_schema)
    .load("/Volumes/workspace/default/airbnb_data/people.csv")
)

# Print out the resulting DataFrame structure
df.printSchema()
```

# Spark SQL

## Spark SQL and DataFrames

**Spark SQL** is the SQL interface to the exact same engine that runs DataFrames — both are fully interchangeable.

### DataFrame Registration
* **Temporary views:** `createOrReplaceTempView()` (scoped to the current Spark session)
* **Global temp views:** `createGlobalTempView()` (shared across sessions in the same application)
* Registration allows you to run standard SQL queries directly against DataFrames.

### SQL Query Execution
* Use `spark.sql()` to execute SQL statements from Python code.
* Results are returned as standard DataFrames, enabling SQL and DataFrame operations to compose seamlessly.
* Full SQL support including complex types, expressions, and built-in functions.
