### The SparkSession

Introduced in Spark 2.0 as the unified entry point for all Spark functionality,
replacing the separate SparkContext, SQLContext, HiveContext, and
StreamingContext objects that earlier versions required.


```python
# Importa a classe SparkSession do módulo SQL do PySpark.
from pyspark.sql import SparkSession

# Inicia o padrão Builder, que permite configurar a aplicação Spark passo a passo antes de a criar.
spark = (SparkSession.builder
.appName("MySparkApplication") # Define o nome da sua aplicação Spark (neste caso, "MySparkApplication"). Este nome aparecerá na interface web do Spark (Spark UI) e nos registos (logs) para ajudar a identificar o trabalho.
.config("spark.some.config.option", "some-value") #Define uma propriedade de configuração específica do Spark. Serve para personalizar o comportamento do ambiente (por exemplo, ajustar parâmetros de memória, definições de desempenho ou integrações).
.getOrCreate()) #Verifica se já existe uma SparkSession ativa. Se existir, recupera essa sessão; caso contrário, cria uma nova com as configurações especificadas. Isto evita a criação de múltiplas sessões num mesmo processo.
```
