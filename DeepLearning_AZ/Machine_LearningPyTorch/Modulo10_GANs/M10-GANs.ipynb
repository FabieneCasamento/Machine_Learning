# Aprendizagem na rede neural => Generative Adversarial Networks (GANs)


discriminar, irá analisar as imagens. E ver o qual próxima a imagem é um cachorro ou gato, com exemplo. 

Uma (1) rede neural para gerar a imagem e outra (1) rede neural para avaliar a imagem (discriminatória). 


1) Primeiro parâmetro gerador
Entrada recebe números aleatórios (vetores com 100 ou 200 números) => gerar imagem, avançando, melhora os pesos, e ajusta a imagem. 

2) Discriminador
Discriminador => recebe como parâmetro imagens de cachorros, aprende, e depois recebe outras imagens, para diferenciar as imagens. 

2 problemas de classificação binária. 

### Exemplo passo a passo com acontece.


2 redes neurais, gerador = redes neurais densas ou  convolucionais para imagem (1) e discriminador redes neurais convolucionais para imagem (2)


gerador recebe números aleatórios (gera imagens na primeira rodada, sem saber nada de imagens). Imagem, é conjuto de pixels, matriz de números. 

Passar as imagens para o discriminador, para aprender, mas também teremos que ter a base das imagens do que é cachorro. No discriminatório, temos apenas 1 neurônio de saída, pois é uma classificação binária. Com a função de ativação sigmoide.Com resposta ou 0 ou 1.

Retorna probabilidades. Valor esperado para cachorro é 1,


(números são probabilidades)**

Sem o treinamento. Repropagation, atualização dos pesos. Para aprender o que é cachorro e o que não é cachorro. 

Depois do treinamento no discriminador, para ver as imagens se é ou não do cachorro. 

Abaixo treinamento agora:

compara com 1, e calcula o erro.
classificação da imagem. 
 
(números são erros, probabilidade, ou cálculo dos erros)


Depois do discriminador dar o feedback para o gerador. Inicia-se o processo novamente, 

Duas redes neurais, entradas aleatórias. Fazer o mesmo processo. Imagens criadas parecidas. Fazer o treinamento do discriminador. 

Com saídas de 0 não cachorro, e 1 é cachorro. Com probabilidades. 

Backpropagation, na rede neural discriminador. Com várias épocas, o que é cachorro ou não. 



Versão 2, depois que a rede neural discriminatória é treinada;

indicando a probabilidade, comparação com 1

pegar os resultados e dará um feedback na rede neural gerador, para gerar novas imagens. 



Verificar as imagens, fazendo várias épocas.
