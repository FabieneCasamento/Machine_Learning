🔹 1. Comparar linha inteira (True/False por linha)

```python
import pandas as pd

# Exemplo
df1 = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
df2 = pd.DataFrame({"A": [1, 9, 3], "B": [4, 5, 0]})

# Comparação linha por linha
comparacao = (df1 == df2).all(axis=1)
print(comparacao)
```

🔹 2. Comparar célula por célula

Se você quiser ver onde exatamente estão as diferenças:

```python
diferencas = df1 != df2
display(diferencas)
```

🔹 3. Usar DataFrame.equals (comparação total)

```python
print(df1.equals(df2))  # retorna True ou False
```

🔹 4. Obter linhas diferentes

```python
linhas_diferentes = df1[~(df1 == df2).all(axis=1)]
print(linhas_diferentes)
```
